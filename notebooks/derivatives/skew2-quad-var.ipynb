```python
df = pd.read_hdf('tick_data.hdf')
df.head()
```

```python
df = df.resample('5T').last().dropna()
```

```python
df['epoch'] = df.index
df['time_of_day'] = df.epoch.dt.floor('10T') - df.epoch.dt.normalize()
start = pd.Timedelta(hours=15, minutes=50)
end = pd.Timedelta(hours=17, minutes=10)
df = df[~df.time_of_day.between(start,end)]
df['mid'] = df.bid_px.add(df.ask_px).div(2)
```
