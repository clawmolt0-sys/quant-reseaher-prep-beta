## 1. Counting and Combinatorics

Counting problems are among the most common in quant interviews. The key is to translate a word problem into a structured counting argument. Mistakes usually arise from overcounting or undercounting --- always ask yourself whether **order matters** and whether **repetition is allowed**.

### 1.1 The Fundamental Counting Principle

If a procedure can be broken into $k$ sequential stages, where stage $i$ has $n_i$ choices (independent of choices made in other stages), then the total number of outcomes is:

$$n_1 \times n_2 \times \cdots \times n_k$$

**Example:** A license plate has 3 letters followed by 4 digits. How many distinct plates are possible?

$$26^3 \times 10^4 = 17{,}576 \times 10{,}000 = 175{,}760{,}000$$

### 1.2 Permutations

A **permutation** is an ordered arrangement. The number of ways to choose and arrange $k$ objects from $n$ distinct objects is:

$$P(n,k) = \frac{n!}{(n-k)!}$$

Special case: the number of ways to arrange all $n$ objects is $n!$.

**Permutations with repetition:** If we have $n$ objects where there are $n_1$ identical objects of type 1, $n_2$ of type 2, ..., $n_r$ of type $r$, then the number of distinct permutations is the **multinomial coefficient**:

$$\frac{n!}{n_1! \, n_2! \cdots n_r!}$$

> **Example:** How many distinct arrangements of the letters in MISSISSIPPI?
>
> We have 11 letters: M(1), I(4), S(4), P(2).
>
> $$\frac{11!}{1! \cdot 4! \cdot 4! \cdot 2!} = \frac{39916800}{1 \cdot 24 \cdot 24 \cdot 2} = 34{,}650$$

### 1.3 Combinations

A **combination** is an unordered selection. The number of ways to choose $k$ objects from $n$ distinct objects (without regard to order) is:

$$\binom{n}{k} = \frac{n!}{k!(n-k)!}$$

Key properties:
- **Symmetry:** $\binom{n}{k} = \binom{n}{n-k}$
- **Pascal's identity:** $\binom{n}{k} = \binom{n-1}{k-1} + \binom{n-1}{k}$
- **Binomial theorem:** $(x+y)^n = \sum_{k=0}^{n} \binom{n}{k} x^k y^{n-k}$
- **Vandermonde's identity:** $\binom{m+n}{r} = \sum_{k=0}^{r} \binom{m}{k}\binom{n}{r-k}$
- **Sum of all:** $\sum_{k=0}^{n}\binom{n}{k} = 2^n$

> **Interview Tip:** When an interviewer asks a counting question, always clarify whether order matters. Saying "I want to confirm --- are we counting ordered or unordered selections?" shows mathematical maturity.

### 1.4 Stars and Bars

**Stars and bars** is a technique for counting the number of ways to distribute $n$ identical objects into $k$ distinct bins.

- **Non-negative integer solutions** to $x_1 + x_2 + \cdots + x_k = n$:

$$\binom{n+k-1}{k-1}$$

- **Positive integer solutions** (each $x_i \geq 1$): substitute $y_i = x_i - 1$ so $y_1 + \cdots + y_k = n - k$:

$$\binom{n-1}{k-1}$$

> **Example:** How many ways can you distribute 10 identical candies among 4 children?
>
> This is the number of non-negative integer solutions to $x_1 + x_2 + x_3 + x_4 = 10$:
>
> $$\binom{10 + 4 - 1}{4-1} = \binom{13}{3} = 286$$

### 1.5 Inclusion-Exclusion Principle

For events $A_1, A_2, \ldots, A_n$:

$$P\!\left(\bigcup_{i=1}^n A_i\right) = \sum_{i} P(A_i) - \sum_{i<j} P(A_i \cap A_j) + \sum_{i<j<k} P(A_i \cap A_j \cap A_k) - \cdots + (-1)^{n+1} P(A_1 \cap \cdots \cap A_n)$$

For two events this simplifies to the familiar:

$$P(A \cup B) = P(A) + P(B) - P(A \cap B)$$

> **Example (Derangements):** A classic application is counting **derangements** --- permutations where no element appears in its original position. The number of derangements of $n$ elements is:
>
> $$D_n = n! \sum_{k=0}^{n} \frac{(-1)^k}{k!} \approx \frac{n!}{e}$$
>
> The probability that a random permutation is a derangement converges to $1/e \approx 0.3679$ as $n \to \infty$.

> **Interview Tip:** Derangement problems are extremely popular in quant interviews. If you are asked "what is the probability that no one gets their own hat back?" --- the answer is approximately $1/e$ for large $n$.

## 2. Probability Axioms and Basic Rules

A rigorous understanding of the probability axioms is essential. This is the mathematical scaffolding on which everything else rests.

### 2.1 Sample Space, Events, and Sigma-Algebras

- **Sample space** $\Omega$: the set of all possible outcomes of a random experiment.
- **Event**: a subset of $\Omega$.
- **Sigma-algebra** $\mathcal{F}$: a collection of subsets of $\Omega$ that is closed under complementation and countable unions. It represents all events to which we can assign probabilities.

A **probability measure** is a function $P: \mathcal{F} \to [0,1]$ satisfying the Kolmogorov axioms.

### 2.2 Kolmogorov Axioms

1. **Non-negativity:** $P(A) \geq 0$ for every event $A \in \mathcal{F}$.
2. **Normalization:** $P(\Omega) = 1$.
3. **Countable additivity:** If $A_1, A_2, \ldots$ are pairwise disjoint events, then:

$$P\!\left(\bigcup_{i=1}^{\infty} A_i\right) = \sum_{i=1}^{\infty} P(A_i)$$

From these three axioms, all of probability theory follows.

### 2.3 Fundamental Rules

**Complement rule:**
$$P(A^c) = 1 - P(A)$$

**Addition rule (general):**
$$P(A \cup B) = P(A) + P(B) - P(A \cap B)$$

**Monotonicity:** If $A \subseteq B$, then $P(A) \leq P(B)$.

**Boole's inequality (union bound):**
$$P\!\left(\bigcup_{i=1}^n A_i\right) \leq \sum_{i=1}^n P(A_i)$$

**Partition theorem (law of total probability):** If $B_1, B_2, \ldots, B_n$ form a partition of $\Omega$ (i.e., they are mutually exclusive and exhaustive), then for any event $A$:

$$P(A) = \sum_{i=1}^{n} P(A \mid B_i)\, P(B_i)$$

### 2.4 Conditional Probability

The **conditional probability** of $A$ given $B$ (where $P(B) > 0$) is:

$$P(A \mid B) = \frac{P(A \cap B)}{P(B)}$$

**Multiplication rule:**
$$P(A \cap B) = P(A \mid B)\, P(B) = P(B \mid A)\, P(A)$$

**Chain rule (generalized):**
$$P(A_1 \cap A_2 \cap \cdots \cap A_n) = P(A_1)\, P(A_2 \mid A_1)\, P(A_3 \mid A_1 \cap A_2) \cdots P(A_n \mid A_1 \cap \cdots \cap A_{n-1})$$

> **Interview Tip:** Many probability puzzles are best solved by conditioning. When stuck, ask yourself: "What is the natural thing to condition on?" This single strategy solves a huge fraction of interview problems.

### 2.5 Independence

Events $A$ and $B$ are **independent** if and only if:

$$P(A \cap B) = P(A)\, P(B)$$

Equivalently, $P(A \mid B) = P(A)$ (knowing $B$ occurred gives no information about $A$).

**Pairwise vs. mutual independence:** For three events $A, B, C$:
- **Pairwise independence** requires: $P(A \cap B) = P(A)P(B)$, $P(A \cap C) = P(A)P(C)$, $P(B \cap C) = P(B)P(C)$.
- **Mutual independence** additionally requires: $P(A \cap B \cap C) = P(A)P(B)P(C)$.

Pairwise independence does **not** imply mutual independence.

> **Interview Tip:** A common trick question: "If $A$ and $B$ are independent, and $B$ and $C$ are independent, are $A$ and $C$ independent?" The answer is **no** --- independence does not propagate transitively.

## 3. Bayes' Theorem

Bayes' theorem is arguably the single most important tool in applied probability. It tells us how to **update beliefs** in light of new evidence.

### 3.1 Statement of Bayes' Theorem

$$P(A \mid B) = \frac{P(B \mid A)\, P(A)}{P(B)}$$

Using the law of total probability to expand the denominator:

$$P(A \mid B) = \frac{P(B \mid A)\, P(A)}{P(B \mid A)\, P(A) + P(B \mid A^c)\, P(A^c)}$$

More generally, if $\{H_1, H_2, \ldots, H_n\}$ is a partition of $\Omega$:

$$P(H_i \mid B) = \frac{P(B \mid H_i)\, P(H_i)}{\sum_{j=1}^{n} P(B \mid H_j)\, P(H_j)}$$

**Terminology:**
- $P(H_i)$: **prior** probability (before observing evidence)
- $P(H_i \mid B)$: **posterior** probability (after observing evidence)
- $P(B \mid H_i)$: **likelihood** of the evidence under hypothesis $H_i$

### 3.2 Classic Example: Medical Testing

> **Example:** A rare disease affects 1 in 10,000 people. A test for the disease has a 99% true positive rate (sensitivity) and a 99% true negative rate (specificity). If a person tests positive, what is the probability they actually have the disease?

Let $D$ = has disease, $T^+$ = tests positive.

**Given:**
- $P(D) = 0.0001$
- $P(T^+ \mid D) = 0.99$
- $P(T^+ \mid D^c) = 0.01$

**Apply Bayes' theorem:**

$$P(D \mid T^+) = \frac{P(T^+ \mid D)\, P(D)}{P(T^+ \mid D)\, P(D) + P(T^+ \mid D^c)\, P(D^c)}$$

$$= \frac{0.99 \times 0.0001}{0.99 \times 0.0001 + 0.01 \times 0.9999}$$

$$= \frac{0.000099}{0.000099 + 0.009999} = \frac{0.000099}{0.010098} \approx 0.0098$$

**Less than 1%!** Despite the test being 99% accurate, the low base rate means most positive results are false positives.

> **Interview Tip:** This is perhaps the most classic Bayes' theorem interview question. The key insight is the **base rate fallacy**: when the prior is very small, even a highly accurate test produces mostly false positives. Always sanity-check your answer by thinking about the relative sizes of the groups.

### 3.3 Odds Form of Bayes' Theorem

An elegant alternative formulation uses **odds**. Define the prior odds as $O(A) = \frac{P(A)}{P(A^c)}$ and the **Bayes factor** as $\text{BF} = \frac{P(B \mid A)}{P(B \mid A^c)}$. Then:

$$\text{Posterior odds} = \text{Bayes factor} \times \text{Prior odds}$$

$$\frac{P(A \mid B)}{P(A^c \mid B)} = \frac{P(B \mid A)}{P(B \mid A^c)} \cdot \frac{P(A)}{P(A^c)}$$

This form is especially useful for sequential updating: each new piece of evidence simply multiplies the current odds by a new Bayes factor.

## 4. Random Variables and Distributions

A **random variable** is a measurable function $X: \Omega \to \mathbb{R}$ that maps outcomes in the sample space to real numbers. Random variables come in two main flavors: **discrete** (taking countably many values) and **continuous** (taking uncountably many values).

### 4.1 Probability Mass Function (PMF)

For a discrete random variable $X$, the **PMF** is:

$$p_X(x) = P(X = x)$$

Properties:
- $p_X(x) \geq 0$ for all $x$
- $\sum_{x} p_X(x) = 1$

### 4.2 Probability Density Function (PDF)

For a continuous random variable $X$, the **PDF** $f_X(x)$ satisfies:

$$P(a \leq X \leq b) = \int_a^b f_X(x)\, dx$$

Properties:
- $f_X(x) \geq 0$ for all $x$
- $\int_{-\infty}^{\infty} f_X(x)\, dx = 1$
- **Important:** $f_X(x)$ is *not* a probability. It can exceed 1. For example, $\text{Uniform}(0, 0.5)$ has $f(x) = 2$ on its support.
- $P(X = x) = 0$ for any specific value $x$.

### 4.3 Cumulative Distribution Function (CDF)

The **CDF** is defined for any random variable (discrete or continuous):

$$F_X(x) = P(X \leq x)$$

Properties:
- $F_X$ is non-decreasing: $x_1 < x_2 \implies F_X(x_1) \leq F_X(x_2)$
- $\lim_{x \to -\infty} F_X(x) = 0$ and $\lim_{x \to \infty} F_X(x) = 1$
- $F_X$ is right-continuous
- For continuous r.v.s: $f_X(x) = F_X'(x)$
- $P(a < X \leq b) = F_X(b) - F_X(a)$

### 4.4 Key Discrete Distributions

| Distribution | PMF | Mean | Variance | MGF |
|:---|:---|:---|:---|:---|
| $\text{Bernoulli}(p)$ | $p^x(1-p)^{1-x}$, $x \in \{0,1\}$ | $p$ | $p(1-p)$ | $(1-p)+pe^t$ |
| $\text{Binomial}(n,p)$ | $\binom{n}{x}p^x(1-p)^{n-x}$ | $np$ | $np(1-p)$ | $((1-p)+pe^t)^n$ |
| $\text{Geometric}(p)$ | $(1-p)^{x-1}p$, $x \geq 1$ | $1/p$ | $(1-p)/p^2$ | $\frac{pe^t}{1-(1-p)e^t}$ |
| $\text{Poisson}(\lambda)$ | $\frac{e^{-\lambda}\lambda^x}{x!}$, $x \geq 0$ | $\lambda$ | $\lambda$ | $e^{\lambda(e^t - 1)}$ |
| $\text{NegBin}(r,p)$ | $\binom{x-1}{r-1}p^r(1-p)^{x-r}$ | $r/p$ | $r(1-p)/p^2$ | $\left(\frac{pe^t}{1-(1-p)e^t}\right)^r$ |

> **Interview Tip:** Know the mean and variance of each distribution cold. Interviewers often ask you to derive them quickly --- the fastest route is usually via the MGF or by using indicator random variables and linearity of expectation.

### 4.5 Key Continuous Distributions

| Distribution | PDF | Mean | Variance | MGF |
|:---|:---|:---|:---|:---|
| $\text{Uniform}(a,b)$ | $\frac{1}{b-a}$, $x \in [a,b]$ | $\frac{a+b}{2}$ | $\frac{(b-a)^2}{12}$ | $\frac{e^{tb}-e^{ta}}{t(b-a)}$ |
| $\text{Exponential}(\lambda)$ | $\lambda e^{-\lambda x}$, $x \geq 0$ | $1/\lambda$ | $1/\lambda^2$ | $\frac{\lambda}{\lambda - t}$, $t < \lambda$ |
| $\text{Normal}(\mu, \sigma^2)$ | $\frac{1}{\sigma\sqrt{2\pi}}e^{-\frac{(x-\mu)^2}{2\sigma^2}}$ | $\mu$ | $\sigma^2$ | $e^{\mu t + \frac{1}{2}\sigma^2 t^2}$ |
| $\text{Gamma}(\alpha,\beta)$ | $\frac{\beta^\alpha}{\Gamma(\alpha)}x^{\alpha-1}e^{-\beta x}$ | $\alpha/\beta$ | $\alpha/\beta^2$ | $\left(\frac{\beta}{\beta-t}\right)^\alpha$ |
| $\text{Beta}(\alpha,\beta)$ | $\frac{x^{\alpha-1}(1-x)^{\beta-1}}{B(\alpha,\beta)}$, $x \in [0,1]$ | $\frac{\alpha}{\alpha+\beta}$ | $\frac{\alpha\beta}{(\alpha+\beta)^2(\alpha+\beta+1)}$ | (no closed form) |

## 5. Expectation and Variance

Expectation and variance are the two most fundamental summary statistics of a distribution. Mastering their properties --- especially **linearity of expectation** --- is critical for quant interviews.

### 5.1 Expected Value

**Discrete case:**
$$E[X] = \sum_{x} x \cdot P(X = x)$$

**Continuous case:**
$$E[X] = \int_{-\infty}^{\infty} x \, f_X(x)\, dx$$

**Law of the Unconscious Statistician (LOTUS):** For any function $g$:
$$E[g(X)] = \sum_x g(x) \cdot P(X=x) \quad \text{or} \quad E[g(X)] = \int_{-\infty}^{\infty} g(x) f_X(x)\, dx$$

You do **not** need to find the distribution of $g(X)$ first.

### 5.2 Linearity of Expectation

This is perhaps the single most powerful tool in discrete probability:

$$E[aX + bY] = a\,E[X] + b\,E[Y]$$

This holds **regardless of whether $X$ and $Y$ are independent**. It extends to any finite sum:

$$E\!\left[\sum_{i=1}^n a_i X_i\right] = \sum_{i=1}^n a_i\, E[X_i]$$

> **Example (Expected number of fixed points):** A random permutation of $n$ elements --- what is the expected number of fixed points (elements that remain in their original position)?
>
> Let $X_i = \mathbf{1}\{\text{element } i \text{ is a fixed point}\}$. Then $E[X_i] = 1/n$ for each $i$.
>
> By linearity: $E\!\left[\sum_{i=1}^n X_i\right] = \sum_{i=1}^n \frac{1}{n} = 1$.
>
> The expected number of fixed points is exactly **1**, regardless of $n$!

> **Interview Tip:** Linearity of expectation is the secret weapon for quant interviews. The strategy is to decompose a complicated random variable into a sum of simple indicator random variables $X = \sum X_i$ where each $X_i$ is easy to analyze. The $X_i$ can be dependent --- it does not matter.

### 5.3 Variance

$$\text{Var}(X) = E[(X - E[X])^2] = E[X^2] - (E[X])^2$$

The second form ($E[X^2] - (E[X])^2$) is almost always the faster way to compute variance.

**Standard deviation:** $\text{SD}(X) = \sigma_X = \sqrt{\text{Var}(X)}$

**Properties:**
- $\text{Var}(aX + b) = a^2 \text{Var}(X)$ (constants shift but don't affect spread; scaling squares)
- $\text{Var}(X) \geq 0$, with equality iff $X$ is constant a.s.

### 5.4 Covariance and Correlation

**Covariance:**
$$\text{Cov}(X, Y) = E[(X - E[X])(Y - E[Y])] = E[XY] - E[X]\,E[Y]$$

**Properties of covariance:**
- $\text{Cov}(X, X) = \text{Var}(X)$
- $\text{Cov}(X, Y) = \text{Cov}(Y, X)$ (symmetric)
- $\text{Cov}(aX, bY) = ab\,\text{Cov}(X, Y)$
- **Bilinearity:** $\text{Cov}(X + Y, Z) = \text{Cov}(X, Z) + \text{Cov}(Y, Z)$
- If $X, Y$ are independent, then $\text{Cov}(X,Y) = 0$ (but the converse is **false**!)

**Correlation:**
$$\rho(X, Y) = \frac{\text{Cov}(X, Y)}{\text{SD}(X) \cdot \text{SD}(Y)}, \qquad -1 \leq \rho \leq 1$$

$|\rho| = 1$ iff $Y = aX + b$ for some constants $a \neq 0, b$ (perfect linear relationship).

### 5.5 Variance of Sums

**General formula:**

$$\text{Var}(aX + bY) = a^2\text{Var}(X) + b^2\text{Var}(Y) + 2ab\,\text{Cov}(X, Y)$$

For $n$ random variables:

$$\text{Var}\!\left(\sum_{i=1}^n X_i\right) = \sum_{i=1}^n \text{Var}(X_i) + 2\sum_{i<j} \text{Cov}(X_i, X_j)$$

**If the $X_i$ are pairwise uncorrelated** (which holds when they are independent):

$$\text{Var}\!\left(\sum_{i=1}^n X_i\right) = \sum_{i=1}^n \text{Var}(X_i)$$

> **Interview Tip:** In finance, the variance of a portfolio return $R_p = \sum w_i R_i$ is $\text{Var}(R_p) = \sum_i \sum_j w_i w_j \text{Cov}(R_i, R_j)$. Understanding this formula and how diversification reduces variance (when correlations are less than 1) is fundamental.

## 6. Common Distributions Deep Dive

This section covers the properties of key distributions in greater depth, focusing on results that frequently appear in quant interviews.

### 6.1 The Normal Distribution

If $X \sim N(\mu, \sigma^2)$, then its PDF is:

$$f(x) = \frac{1}{\sigma\sqrt{2\pi}} \exp\!\left(-\frac{(x-\mu)^2}{2\sigma^2}\right)$$

**Standardization:** If $X \sim N(\mu, \sigma^2)$, then $Z = \frac{X - \mu}{\sigma} \sim N(0,1)$.

**The 68-95-99.7 rule:**
- $P(\mu - \sigma < X < \mu + \sigma) \approx 0.6827$
- $P(\mu - 2\sigma < X < \mu + 2\sigma) \approx 0.9545$
- $P(\mu - 3\sigma < X < \mu + 3\sigma) \approx 0.9973$

**Key properties:**

1. **Linear combinations are normal:** If $X \sim N(\mu_X, \sigma_X^2)$ and $Y \sim N(\mu_Y, \sigma_Y^2)$ are independent, then:
$$aX + bY \sim N(a\mu_X + b\mu_Y,\; a^2\sigma_X^2 + b^2\sigma_Y^2)$$

2. **Sum of $n$ i.i.d. normals:** If $X_i \sim N(\mu, \sigma^2)$ i.i.d., then:
$$\sum_{i=1}^n X_i \sim N(n\mu,\; n\sigma^2), \qquad \bar{X} \sim N\!\left(\mu,\; \frac{\sigma^2}{n}\right)$$

3. **MGF:** $M_X(t) = e^{\mu t + \frac{1}{2}\sigma^2 t^2}$

4. **Moments of the standard normal $Z \sim N(0,1)$:**
   - All odd moments are 0 (by symmetry)
   - $E[Z^{2n}] = (2n-1)!! = 1 \cdot 3 \cdot 5 \cdots (2n-1)$
   - In particular: $E[Z^2] = 1$, $E[Z^4] = 3$, $E[Z^6] = 15$

5. **The normal CDF** $\Phi(z) = P(Z \leq z)$ has no closed form but key values to memorize:
   - $\Phi(0) = 0.5$, $\Phi(1) \approx 0.8413$, $\Phi(1.645) \approx 0.95$, $\Phi(1.96) \approx 0.975$, $\Phi(2.326) \approx 0.99$

### 6.2 The Poisson Distribution and Poisson Process

If $X \sim \text{Poisson}(\lambda)$:

$$P(X = k) = \frac{e^{-\lambda} \lambda^k}{k!}, \qquad k = 0, 1, 2, \ldots$$

- $E[X] = \lambda$, $\text{Var}(X) = \lambda$ (mean equals variance --- a key identifying feature).

**Poisson as a limit of Binomial:** If $X \sim \text{Binomial}(n, p)$ with $n \to \infty$, $p \to 0$, and $np \to \lambda$, then $X \xrightarrow{d} \text{Poisson}(\lambda)$.

**Sum of Poissons:** If $X \sim \text{Poisson}(\lambda)$ and $Y \sim \text{Poisson}(\mu)$ are independent, then $X + Y \sim \text{Poisson}(\lambda + \mu)$.

**Poisson process:** A counting process $\{N(t), t \geq 0\}$ is a Poisson process with rate $\lambda$ if:
- $N(0) = 0$
- It has independent increments
- $N(t) - N(s) \sim \text{Poisson}(\lambda(t-s))$ for $0 \leq s < t$

The inter-arrival times in a Poisson process are i.i.d. $\text{Exponential}(\lambda)$.

### 6.3 The Exponential Distribution and Memoryless Property

If $X \sim \text{Exponential}(\lambda)$:

$$f(x) = \lambda e^{-\lambda x}, \quad x \geq 0 \qquad F(x) = 1 - e^{-\lambda x}$$

- $E[X] = 1/\lambda$, $\text{Var}(X) = 1/\lambda^2$.

**The memoryless property:**

$$P(X > s + t \mid X > s) = P(X > t) \quad \text{for all } s, t \geq 0$$

The exponential distribution is the **only** continuous distribution with this property. (For discrete distributions, the geometric distribution is the unique memoryless distribution.)

**Minimum of exponentials:** If $X_1, \ldots, X_n$ are independent with $X_i \sim \text{Exp}(\lambda_i)$, then:
$$\min(X_1, \ldots, X_n) \sim \text{Exp}(\lambda_1 + \lambda_2 + \cdots + \lambda_n)$$

**Connection to Poisson:** In a Poisson process with rate $\lambda$, the time until the first event is $\text{Exp}(\lambda)$, and the time until the $n$-th event is $\text{Gamma}(n, \lambda)$.

> **Interview Tip:** The memoryless property is a staple of quant interviews. A classic question: "You arrive at a bus stop where buses come according to a Poisson process with rate $\lambda$. What is your expected waiting time?" Answer: $1/\lambda$, by the memoryless property --- it does not matter when the last bus came.

## 7. Conditional Expectation and the Law of Total Expectation

Conditional expectation is one of the most powerful tools in probability and forms the backbone of stochastic calculus, which in turn underlies derivatives pricing.

### 7.1 Conditional Expectation

**Discrete case:** If $X$ and $Y$ are discrete:
$$E[X \mid Y = y] = \sum_x x \cdot P(X = x \mid Y = y)$$

**Continuous case:**
$$E[X \mid Y = y] = \int_{-\infty}^{\infty} x \cdot f_{X|Y}(x \mid y)\, dx$$

Note that $E[X \mid Y = y]$ is a function of $y$, while $E[X \mid Y]$ (without specifying the value) is a **random variable** --- it is a function of $Y$.

### 7.2 Law of Total Expectation (Adam's Law / Tower Property)

$$E[X] = E[E[X \mid Y]]$$

In the discrete case this becomes:
$$E[X] = \sum_y E[X \mid Y = y] \cdot P(Y = y)$$

This is also called the **tower property** or **law of iterated expectations**. In more generality:

$$E[E[X \mid \mathcal{G}] \mid \mathcal{H}] = E[X \mid \mathcal{H}] \quad \text{when } \mathcal{H} \subseteq \mathcal{G}$$

That is, conditioning on less information "smooths out" the inner conditioning.

> **Example:** What is the expected number of coin flips to get heads, if the coin's probability $p$ of heads is itself random, drawn from $\text{Uniform}(0,1)$?
>
> Conditional on $p$, the number of flips is $\text{Geometric}(p)$ with mean $1/p$. So:
>
> $$E[X] = E[E[X \mid p]] = E[1/p] = \int_0^1 \frac{1}{p} \cdot 1 \, dp = \infty$$
>
> The expected number of flips is infinite!

### 7.3 Law of Total Variance (Eve's Law)

$$\text{Var}(X) = E[\text{Var}(X \mid Y)] + \text{Var}(E[X \mid Y])$$

**Interpretation:**
- $E[\text{Var}(X \mid Y)]$: the average "within-group" variance (unexplained variance)
- $\text{Var}(E[X \mid Y])$: the variance "between groups" (explained variance)

Total variance = average of conditional variances + variance of conditional means.

**Mnemonic:** "Eve's Law" because $E$ of $\text{Var}$ + $\text{Var}$ of $E$ (E-V-V-E).

> **Example:** Roll a die to get $N$. Then flip $N$ fair coins. Let $X$ = number of heads. Find $\text{Var}(X)$.
>
> Given $N = n$: $X \mid N = n \sim \text{Binomial}(n, 1/2)$, so $E[X \mid N] = N/2$ and $\text{Var}(X \mid N) = N/4$.
>
> $E[N] = 7/2$, $E[N^2] = 91/6$, $\text{Var}(N) = 35/12$.
>
> By Eve's law:
> $$\text{Var}(X) = E[N/4] + \text{Var}(N/2) = \frac{E[N]}{4} + \frac{\text{Var}(N)}{4} = \frac{7/2}{4} + \frac{35/12}{4} = \frac{7}{8} + \frac{35}{48} = \frac{42 + 35}{48} = \frac{77}{48}$$

## 8. Moment Generating Functions

Moment generating functions (MGFs) are a powerful tool for characterizing distributions, computing moments, and proving convergence results.

### 8.1 Definition and Properties

The **moment generating function** of a random variable $X$ is:

$$M_X(t) = E[e^{tX}]$$

defined for all $t$ in some neighborhood of 0.

**Why "moment generating"?** The $n$-th moment of $X$ is:

$$E[X^n] = M_X^{(n)}(0) = \frac{d^n}{dt^n} M_X(t)\Big|_{t=0}$$

**Key properties:**

1. **Uniqueness:** If the MGF exists in a neighborhood of 0, it uniquely determines the distribution.

2. **Linear transformation:** If $Y = aX + b$, then $M_Y(t) = e^{bt} M_X(at)$.

3. **Sum of independents:** If $X$ and $Y$ are independent, then $M_{X+Y}(t) = M_X(t) \cdot M_Y(t)$.

4. **Convergence:** If $M_{X_n}(t) \to M_X(t)$ for all $t$ in a neighborhood of 0, then $X_n \xrightarrow{d} X$.

### 8.2 Computing Moments via MGF

**Example: Poisson distribution.**

If $X \sim \text{Poisson}(\lambda)$, then $M_X(t) = e^{\lambda(e^t - 1)}$.

First moment:
$$M_X'(t) = \lambda e^t \cdot e^{\lambda(e^t - 1)}$$
$$E[X] = M_X'(0) = \lambda \cdot 1 = \lambda$$

Second moment:
$$M_X''(t) = (\lambda e^t + \lambda^2 e^{2t}) \cdot e^{\lambda(e^t - 1)}$$
$$E[X^2] = M_X''(0) = \lambda + \lambda^2$$

Therefore:
$$\text{Var}(X) = E[X^2] - (E[X])^2 = (\lambda + \lambda^2) - \lambda^2 = \lambda \; \checkmark$$

### 8.3 MGFs of Common Distributions (Summary)

| Distribution | MGF $M_X(t)$ | Domain |
|:---|:---|:---|
| $\text{Bernoulli}(p)$ | $(1-p) + pe^t$ | all $t$ |
| $\text{Binomial}(n,p)$ | $((1-p) + pe^t)^n$ | all $t$ |
| $\text{Poisson}(\lambda)$ | $e^{\lambda(e^t-1)}$ | all $t$ |
| $\text{Geometric}(p)$ | $\frac{pe^t}{1-(1-p)e^t}$ | $t < -\ln(1-p)$ |
| $\text{Exponential}(\lambda)$ | $\frac{\lambda}{\lambda - t}$ | $t < \lambda$ |
| $\text{Normal}(\mu, \sigma^2)$ | $e^{\mu t + \frac{1}{2}\sigma^2 t^2}$ | all $t$ |
| $\text{Gamma}(\alpha, \beta)$ | $\left(\frac{\beta}{\beta - t}\right)^\alpha$ | $t < \beta$ |
| $\text{Chi-squared}(k)$ | $(1 - 2t)^{-k/2}$ | $t < 1/2$ |

> **Interview Tip:** MGFs are the fastest way to prove that the sum of independent normals is normal, or that the sum of independent Poissons is Poisson. If two distributions have the same MGF, they are the same distribution. Interviewers love asking: "Prove that $X + Y$ is [some distribution]" --- and the MGF approach is almost always the cleanest.

## 9. Important Inequalities

Probability inequalities let us bound quantities that may be hard to compute exactly. They are also conceptually important in finance, where we often want to bound the probability of extreme losses.

### 9.1 Markov's Inequality

If $X \geq 0$ and $a > 0$:

$$P(X \geq a) \leq \frac{E[X]}{a}$$

**Proof sketch:** $E[X] = E[X \cdot \mathbf{1}_{X \geq a}] + E[X \cdot \mathbf{1}_{X < a}] \geq E[X \cdot \mathbf{1}_{X \geq a}] \geq a \cdot P(X \geq a)$.

Markov's inequality is weak but extremely general --- it only requires the mean to exist.

### 9.2 Chebyshev's Inequality

For any random variable $X$ with finite mean $\mu$ and variance $\sigma^2$, and for any $k > 0$:

$$P(|X - \mu| \geq k\sigma) \leq \frac{1}{k^2}$$

Equivalently, for any $\epsilon > 0$:

$$P(|X - \mu| \geq \epsilon) \leq \frac{\sigma^2}{\epsilon^2}$$

**Proof:** Apply Markov's inequality to $Y = (X - \mu)^2$: $P((X-\mu)^2 \geq \epsilon^2) \leq \frac{E[(X-\mu)^2]}{\epsilon^2} = \frac{\sigma^2}{\epsilon^2}$.

> **Interview Tip:** When asked "what fraction of a distribution can lie more than $k$ standard deviations from the mean?" --- Chebyshev says at most $1/k^2$, regardless of the distribution shape. For $k = 2$: at most 25%. For $k = 3$: at most 11.1%. Compare this with the normal distribution's much tighter bounds.

### 9.3 Jensen's Inequality

If $\varphi$ is a **convex** function and $X$ is a random variable with $E[|X|] < \infty$:

$$\varphi(E[X]) \leq E[\varphi(X)]$$

If $\varphi$ is **concave**, the inequality reverses:

$$\varphi(E[X]) \geq E[\varphi(X)]$$

**Applications in finance:**

1. Since $x^2$ is convex: $E[X^2] \geq (E[X])^2$, which is equivalent to $\text{Var}(X) \geq 0$.

2. Since $\ln(x)$ is concave: $\ln(E[X]) \geq E[\ln(X)]$. This means the **log of the expected return** exceeds the **expected log return** --- this is the foundation of the distinction between arithmetic and geometric mean returns.

3. Since $e^x$ is convex: $E[e^X] \geq e^{E[X]}$. This is why the expected value of a lognormal random variable exceeds $e^{\mu}$.

> **Interview Tip:** Jensen's inequality is extremely important in quantitative finance. It explains why "volatility drag" reduces compound returns, why options have time value, and why risk-neutral pricing requires a change of measure. If an interviewer asks about the difference between arithmetic and geometric mean returns, Jensen's inequality is the answer.

### 9.4 Cauchy-Schwarz Inequality

For any random variables $X$ and $Y$ with finite second moments:

$$|E[XY]| \leq \sqrt{E[X^2] \cdot E[Y^2]}$$

Or equivalently:

$$(E[XY])^2 \leq E[X^2] \cdot E[Y^2]$$

**Corollary:** $|\text{Cov}(X,Y)| \leq \text{SD}(X) \cdot \text{SD}(Y)$, which is why $|\rho(X,Y)| \leq 1$.

Equality holds iff $Y = cX$ for some constant $c$ (a.s.).

## 10. Limit Theorems

The limit theorems are the crown jewels of probability theory. They explain why averages are predictable and why the normal distribution appears everywhere.

### 10.1 Modes of Convergence

Before stating the limit theorems, we need the different notions of convergence:

1. **Almost sure (a.s.) convergence:** $X_n \xrightarrow{a.s.} X$ means $P(\lim_{n\to\infty} X_n = X) = 1$.

2. **Convergence in probability:** $X_n \xrightarrow{P} X$ means $\forall \epsilon > 0$: $P(|X_n - X| > \epsilon) \to 0$.

3. **Convergence in distribution:** $X_n \xrightarrow{d} X$ means $F_{X_n}(x) \to F_X(x)$ at all continuity points of $F_X$.

**Implications:** a.s. $\Rightarrow$ in probability $\Rightarrow$ in distribution (but not the reverse, in general).

### 10.2 Law of Large Numbers

Let $X_1, X_2, \ldots$ be i.i.d. with $E[X_i] = \mu$ and define $\bar{X}_n = \frac{1}{n}\sum_{i=1}^n X_i$.

**Weak Law of Large Numbers (WLLN):** $\bar{X}_n \xrightarrow{P} \mu$ as $n \to \infty$.

*Requires only that $E[X_i] = \mu$ exists.* A simple proof uses Chebyshev's inequality when $\text{Var}(X_i) < \infty$:

$$P(|\bar{X}_n - \mu| \geq \epsilon) \leq \frac{\text{Var}(\bar{X}_n)}{\epsilon^2} = \frac{\sigma^2}{n\epsilon^2} \to 0$$

**Strong Law of Large Numbers (SLLN):** $\bar{X}_n \xrightarrow{a.s.} \mu$ as $n \to \infty$.

This is a stronger statement: the sample mean converges to the population mean with probability 1.

### 10.3 Central Limit Theorem

Let $X_1, X_2, \ldots$ be i.i.d. with $E[X_i] = \mu$ and $\text{Var}(X_i) = \sigma^2 < \infty$. Then:

$$\frac{\bar{X}_n - \mu}{\sigma / \sqrt{n}} = \frac{\sum_{i=1}^n X_i - n\mu}{\sigma\sqrt{n}} \xrightarrow{d} N(0,1) \quad \text{as } n \to \infty$$

Equivalently, for large $n$:

$$\bar{X}_n \approx N\!\left(\mu, \frac{\sigma^2}{n}\right), \qquad \sum_{i=1}^n X_i \approx N(n\mu, n\sigma^2)$$

**Why the CLT matters:** The CLT says that sums of many independent random variables are approximately normal, *regardless of the underlying distribution*. This is why the normal distribution appears everywhere in nature and finance.

**Proof idea (via MGFs):**
Let $Z_n = \frac{\bar{X}_n - \mu}{\sigma/\sqrt{n}}$. The MGF of $Z_n$ is:
$$M_{Z_n}(t) = \left[M_{(X_i-\mu)/\sigma}\!\left(\frac{t}{\sqrt{n}}\right)\right]^n$$

Using the Taylor expansion $M_{(X-\mu)/\sigma}(s) \approx 1 + \frac{s^2}{2} + O(s^3)$ for small $s$:

$$M_{Z_n}(t) \approx \left(1 + \frac{t^2}{2n}\right)^n \to e^{t^2/2}$$

which is the MGF of $N(0,1)$.

### 10.4 Applications in Finance

**Portfolio diversification:** Consider a portfolio of $n$ assets with i.i.d. returns $R_i$ having mean $\mu$ and variance $\sigma^2$. The equally weighted portfolio return is $\bar{R}_n = \frac{1}{n}\sum R_i$.

- $E[\bar{R}_n] = \mu$ (diversification does not change expected return)
- $\text{Var}(\bar{R}_n) = \sigma^2 / n \to 0$ (diversification reduces risk)

The LLN guarantees $\bar{R}_n \to \mu$ and the CLT tells us the distribution is approximately normal for large $n$.

**In practice**, asset returns are **not** independent (they share common risk factors), so diversification cannot eliminate all risk. The residual risk from common factors is called **systematic risk** and is the basis of the CAPM and factor models.

> **Interview Tip:** If asked "why does diversification reduce risk?" the mathematical answer is the law of large numbers. If asked "why can't diversification eliminate all risk?" the answer is that real assets are not independent --- correlated risks cannot be diversified away.

## 11. Classic Interview Problems

The following problems are extremely common in quant finance interviews. Work through each one carefully --- the reasoning patterns here come up again and again.

### Problem 1: Expected Number of Dice Rolls to See All Faces (Coupon Collector)

> **Problem:** You roll a fair 6-sided die repeatedly. What is the expected number of rolls until you have seen all 6 faces at least once?

**Solution:**

Decompose the process into phases. In phase $i$, you have already seen $i-1$ distinct faces and are waiting to see a new one.

- In phase $i$, the probability of rolling a new face is $p_i = \frac{6 - (i-1)}{6} = \frac{7-i}{6}$.
- The number of rolls in phase $i$ is $\text{Geometric}(p_i)$ with mean $\frac{6}{7-i}$.

By linearity of expectation, the total expected number of rolls is:

$$E[T] = \sum_{i=1}^{6} \frac{6}{7-i} = \frac{6}{6} + \frac{6}{5} + \frac{6}{4} + \frac{6}{3} + \frac{6}{2} + \frac{6}{1}$$

$$= 1 + 1.2 + 1.5 + 2 + 3 + 6 = 14.7$$

More generally, for $n$ faces: $E[T] = n \cdot H_n = n \sum_{k=1}^n \frac{1}{k} \approx n \ln n + \gamma n$ where $\gamma \approx 0.5772$ is the Euler-Mascheroni constant.

> **Interview Tip:** This is the **Coupon Collector's Problem**. The key technique is decomposing into geometric random variables and using linearity of expectation. This pattern appears in many variations.

### Problem 2: Breaking a Stick

> **Problem:** A stick of length 1 is broken at two points chosen uniformly at random. What is the probability that the three pieces can form a triangle?

**Solution:**

Let the break points be $U, V \sim \text{Uniform}(0,1)$ independently. Let $X = \min(U,V)$, $Y = \max(U,V)$. The three pieces have lengths $X$, $Y - X$, and $1 - Y$.

Three segments can form a triangle iff the **triangle inequality** holds for all three pairs. Since the lengths sum to 1, this reduces to: **no piece has length $\geq 1/2$**.

We need:
$$X < \frac{1}{2}, \quad Y - X < \frac{1}{2}, \quad 1 - Y < \frac{1}{2}$$

In the $(U,V)$ unit square, the joint density is 1. By symmetry, consider the half where $U < V$ (the other half is symmetric). The constraints become:
- $U < 1/2$
- $V - U < 1/2$, i.e., $V < U + 1/2$
- $V > 1/2$

The area of this region in the triangle $\{0 < U < V < 1\}$ is $1/8$. By symmetry over both halves, the total probability is $2 \times 1/8 = 1/4$.

$$\boxed{P(\text{triangle}) = \frac{1}{4}}$$

### Problem 3: The Gambler's Ruin

> **Problem:** A gambler starts with $\$k$ and repeatedly plays a game where they win $\$1$ with probability $p$ and lose $\$1$ with probability $q = 1-p$. They stop when they reach $\$0$ (ruin) or $\$n$ (target). What is the probability of ruin?

**Solution:**

Let $P_i = P(\text{ruin} \mid \text{starting with } \$i)$. By conditioning on the first step:

$$P_i = p \cdot P_{i+1} + q \cdot P_{i-1}$$

with boundary conditions $P_0 = 1$ (already ruined) and $P_n = 0$ (reached target).

This is a linear recurrence. The solution depends on whether $p = q$:

**Case 1: $p \neq q$ (biased game).** Let $r = q/p$.
$$P_i = \frac{r^i - r^n}{1 - r^n}$$

**Case 2: $p = q = 1/2$ (fair game).**
$$P_i = 1 - \frac{i}{n}$$

Starting with $\$k$, the probability of ruin is:
$$P_k = \begin{cases} \frac{(q/p)^k - (q/p)^n}{1 - (q/p)^n} & p \neq q \\\\ 1 - k/n & p = q = 1/2 \end{cases}$$

> **Interview Tip:** The Gambler's Ruin is a classic random walk problem. The technique of setting up a recurrence by conditioning on the first step is a cornerstone method. Notice that even in the fair game, a gambler with finite wealth will eventually go broke against an infinitely wealthy opponent (let $n \to \infty$: $P_k \to 1$).

### Problem 4: Expected Value of the Maximum of Two Dice

> **Problem:** Roll two fair 6-sided dice. What is the expected value of the maximum of the two rolls?

**Solution:**

Let $M = \max(X_1, X_2)$ where $X_1, X_2$ are independent uniform on $\{1, 2, 3, 4, 5, 6\}$.

**Method 1: CDF approach.**

$$P(M \leq k) = P(X_1 \leq k)\, P(X_2 \leq k) = \left(\frac{k}{6}\right)^2$$

$$P(M = k) = P(M \leq k) - P(M \leq k-1) = \frac{k^2}{36} - \frac{(k-1)^2}{36} = \frac{2k - 1}{36}$$

$$E[M] = \sum_{k=1}^{6} k \cdot \frac{2k-1}{36} = \frac{1}{36}\sum_{k=1}^{6}(2k^2 - k)$$

$$= \frac{1}{36}\left(2 \cdot 91 - 21\right) = \frac{161}{36} \approx 4.472$$

**Method 2: Using the identity** $\max(a,b) = a + b - \min(a,b)$.

By linearity: $E[\max] = E[X_1] + E[X_2] - E[\min]$. By a symmetric CDF argument, $E[\min] = \frac{91}{36}$.

$$E[\max] = 3.5 + 3.5 - \frac{91}{36} = 7 - 2.528 = \frac{161}{36} \approx 4.472$$

### Problem 5: The Monty Hall Problem

> **Problem:** You are on a game show with 3 doors. Behind one door is a car; behind the other two are goats. You pick a door. The host, who knows what is behind each door, opens another door to reveal a goat. You are then given the option to switch to the remaining unopened door. Should you switch?

**Solution using Bayes' theorem:**

Suppose you initially pick Door 1, and the host opens Door 3 (showing a goat). Let $C_i$ = event that the car is behind door $i$.

**Prior:** $P(C_1) = P(C_2) = P(C_3) = 1/3$.

Let $H_3$ = host opens door 3. We need $P(C_2 \mid H_3)$.

**Likelihoods:**
- $P(H_3 \mid C_1) = 1/2$ (car is behind your door; host picks randomly between doors 2 and 3)
- $P(H_3 \mid C_2) = 1$ (car is behind door 2; host must open door 3)
- $P(H_3 \mid C_3) = 0$ (host would never reveal the car)

**By Bayes' theorem:**

$$P(C_2 \mid H_3) = \frac{P(H_3 \mid C_2)\, P(C_2)}{P(H_3)} = \frac{1 \cdot 1/3}{1/2 \cdot 1/3 + 1 \cdot 1/3 + 0 \cdot 1/3} = \frac{1/3}{1/2} = \frac{2}{3}$$

$$P(C_1 \mid H_3) = \frac{P(H_3 \mid C_1)\, P(C_1)}{P(H_3)} = \frac{1/2 \cdot 1/3}{1/2} = \frac{1}{3}$$

**You should always switch!** Switching gives you a $2/3$ probability of winning, while staying gives only $1/3$.

**Intuition:** Your initial choice has a $1/3$ chance of being correct. The host's action does not change this. So the remaining door has a $2/3$ chance.

> **Interview Tip:** The Monty Hall problem tests your ability to reason carefully with conditional probability. The key insight is that the host's action is **informative** because the host has knowledge and is constrained (cannot open the door with the car). Always be precise about what information is being conditioned on.
