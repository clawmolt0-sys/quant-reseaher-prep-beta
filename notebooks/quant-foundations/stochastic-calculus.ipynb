## 1. Brownian Motion

Brownian motion (also called a Wiener process) is the fundamental building block of continuous-time stochastic calculus and the backbone of quantitative finance.

### 1.1 Definition and Construction

> **Definition:** A standard Brownian motion (Wiener process) is a stochastic process $\{B(t)\}_{t \geq 0}$ defined on a probability space $(\Omega, \mathcal{F}, \mathbb{P})$ satisfying:
>
> 1. $B(0) = 0$ almost surely.
> 2. **Independent increments:** For any $0 \leq t_1 < t_2 < \cdots < t_n$, the increments $B(t_2) - B(t_1), B(t_3) - B(t_2), \ldots, B(t_n) - B(t_{n-1})$ are mutually independent.
> 3. **Gaussian increments:** For any $0 \leq s < t$, $B(t) - B(s) \sim \mathcal{N}(0, t - s)$.
> 4. **Continuous paths:** $t \mapsto B(t)$ is continuous almost surely.

From these axioms, the following properties are immediate:

$$
\mathbb{E}[B(t)] = 0, \qquad \text{Var}(B(t)) = t, \qquad \text{Cov}(B(s), B(t)) = \min(s, t)
$$

**Proof of the covariance formula.** Without loss of generality, assume $s \leq t$. Write $B(t) = B(s) + (B(t) - B(s))$. Then:

$$
\text{Cov}(B(s), B(t)) = \mathbb{E}[B(s) \cdot B(t)] = \mathbb{E}[B(s)(B(s) + (B(t) - B(s)))]
$$

$$
= \mathbb{E}[B(s)^2] + \mathbb{E}[B(s)(B(t) - B(s))] = s + 0 = s = \min(s,t)
$$

where we used independence of $B(s)$ and $B(t) - B(s)$.

> **Interview Tip:** When asked about Brownian motion properties, always start with the four defining axioms. Interviewers often follow up with: "Which of these properties is hardest to verify?" The answer is continuity of paths -- the other three are straightforward from the finite-dimensional distributions.

### 1.2 Quadratic Variation

The quadratic variation of Brownian motion is one of the most important properties in stochastic calculus and is what makes Ito's calculus fundamentally different from ordinary calculus.

> **Definition:** The quadratic variation of a process $X$ over $[0, t]$ is:
>
> $$
> [X, X]_t = \lim_{\|\Pi\| \to 0} \sum_{i=0}^{n-1} (X(t_{i+1}) - X(t_i))^2
> $$
>
> where $\Pi = \{0 = t_0 < t_1 < \cdots < t_n = t\}$ is a partition and $\|\Pi\| = \max_i(t_{i+1} - t_i)$.

**Theorem.** For standard Brownian motion, $[B, B]_t = t$.

**Proof sketch.** Consider the sum $Q_n = \sum_{i=0}^{n-1}(B(t_{i+1}) - B(t_i))^2$ for an equally spaced partition with $\Delta t = t/n$. Each term $(B(t_{i+1}) - B(t_i))^2$ has:

$$
\mathbb{E}[(B(t_{i+1}) - B(t_i))^2] = \Delta t, \qquad \text{Var}((B(t_{i+1}) - B(t_i))^2) = 2(\Delta t)^2
$$

The variance of $Q_n$ is (by independence of increments):

$$
\text{Var}(Q_n) = \sum_{i=0}^{n-1} 2(\Delta t)^2 = 2n \cdot \left(\frac{t}{n}\right)^2 = \frac{2t^2}{n} \to 0
$$

Since $\mathbb{E}[Q_n] = t$ and $\text{Var}(Q_n) \to 0$, we have $Q_n \to t$ in $L^2$ (and hence in probability).

The heuristic rule that encapsulates this is:

$$
(dB)^2 = dt, \qquad (dB)(dt) = 0, \qquad (dt)^2 = 0
$$

This "multiplication table" is the key to applying Ito's lemma correctly.

> **Interview Tip:** If asked "Why is stochastic calculus different from ordinary calculus?", the answer is quadratic variation. For smooth (finite variation) functions, quadratic variation is zero, so the second-order term in the Taylor expansion vanishes. For Brownian motion, the quadratic variation is $t$, forcing us to keep the second-order term -- this is exactly the extra $\frac{1}{2}\sigma^2 f''$ term in Ito's lemma.

### 1.3 Path Properties

Brownian motion paths have remarkable and somewhat paradoxical properties:

1. **Continuity:** Paths are almost surely continuous (by definition).
2. **Nowhere differentiable:** Almost surely, $B(t)$ is differentiable at no point $t$. This is a consequence of the quadratic variation being nonzero.
3. **Unbounded variation:** On any interval $[0, t]$, the total variation of $B$ is infinite a.s. This is precisely why we cannot define $\int_0^t f(s) \, dB(s)$ as a Riemann-Stieltjes integral.
4. **Holder continuity:** Brownian paths are almost surely Holder continuous of order $\alpha$ for all $\alpha < 1/2$, but not for $\alpha = 1/2$.
5. **Self-similarity (scaling):** If $B(t)$ is a standard BM, then for any $c > 0$, the process $\tilde{B}(t) = \frac{1}{\sqrt{c}} B(ct)$ is also a standard BM.
6. **Time inversion:** $\tilde{B}(t) = t B(1/t)$ for $t > 0$ (with $\tilde{B}(0) = 0$) is a standard BM.

> **Interview Tip:** A classic interview question is: "Is Brownian motion differentiable?" The answer is no -- it is almost surely nowhere differentiable. Follow up by explaining that this is related to the infinite total variation and nonzero quadratic variation. This is precisely why we need Ito calculus rather than ordinary calculus.

### 1.4 Reflection Principle

The reflection principle is a powerful tool for computing distributions related to the maximum of Brownian motion.

Let $M(t) = \max_{0 \leq s \leq t} B(s)$ be the running maximum. The reflection principle states:

$$
\mathbb{P}(M(t) \geq a) = 2\,\mathbb{P}(B(t) \geq a) = 2\left(1 - \Phi\left(\frac{a}{\sqrt{t}}\right)\right), \qquad a > 0
$$

where $\Phi$ is the standard normal CDF.

**Idea of proof.** By the strong Markov property, after $B$ first hits level $a > 0$ at the stopping time $\tau_a = \inf\{t : B(t) = a\}$, the reflected process $\hat{B}(t) = 2a - B(t)$ for $t \geq \tau_a$ is again a Brownian motion. So for $B(t) \geq a$, the reflected path has $\hat{B}(t) \leq a$ with the same probability, giving:

$$
\mathbb{P}(M(t) \geq a, B(t) \leq a) = \mathbb{P}(M(t) \geq a, B(t) \geq a) = \mathbb{P}(B(t) \geq a)
$$

Adding these:

$$
\mathbb{P}(M(t) \geq a) = 2\,\mathbb{P}(B(t) \geq a)
$$

A useful consequence is the **joint density** of $(B(t), M(t))$:

$$
\mathbb{P}(B(t) \leq x, M(t) \leq m) = \Phi\left(\frac{x}{\sqrt{t}}\right) - \Phi\left(\frac{x - 2m}{\sqrt{t}}\right), \qquad x \leq m, \quad m \geq 0
$$

This is essential for pricing **barrier options**.

> **Interview Tip:** The reflection principle frequently appears in barrier option pricing questions. Be ready to derive the distribution of $M(t) = \max_{s \leq t} B(s)$ and connect it to the probability that Brownian motion hits a barrier before expiry.

## 2. Stochastic Integration

Because Brownian motion paths have infinite variation, the Riemann-Stieltjes integral $\int f \, dB$ does not exist in the classical sense. Ito's construction provides the right framework.

### 2.1 Construction of the Ito Integral

The Ito integral $\int_0^T H(t) \, dB(t)$ is constructed in three steps:

**Step 1: Simple (elementary) processes.** For $H(t) = \sum_{i=0}^{n-1} H_i \mathbf{1}_{(t_i, t_{i+1}]}(t)$, where $H_i$ is $\mathcal{F}_{t_i}$-measurable, define:

$$
\int_0^T H(t) \, dB(t) = \sum_{i=0}^{n-1} H_i (B(t_{i+1}) - B(t_i))
$$

Note the crucial point: $H_i$ is evaluated at the **left** endpoint $t_i$. This non-anticipating (adapted) property is essential.

**Step 2: Ito isometry.** For simple processes, one can verify:

$$
\mathbb{E}\left[\left(\int_0^T H(t) \, dB(t)\right)^2\right] = \mathbb{E}\left[\int_0^T H(t)^2 \, dt\right]
$$

**Proof for simple processes:**

$$
\mathbb{E}\left[\left(\sum_i H_i \Delta B_i\right)^2\right] = \sum_i \mathbb{E}[H_i^2 (\Delta B_i)^2] + 2\sum_{i < j} \mathbb{E}[H_i \Delta B_i H_j \Delta B_j]
$$

The cross terms vanish because $H_j \Delta B_j$ is independent of $\mathcal{F}_{t_j}$ and $\mathbb{E}[\Delta B_j] = 0$. For the diagonal terms, since $H_i$ is $\mathcal{F}_{t_i}$-measurable and $\Delta B_i$ is independent of $\mathcal{F}_{t_i}$:

$$
\mathbb{E}[H_i^2 (\Delta B_i)^2] = \mathbb{E}[H_i^2] \cdot \Delta t_i
$$

Summing gives the result.

**Step 3: Extension by density.** The space of simple processes is dense in $L^2(\Omega \times [0,T])$ (adapted processes with $\mathbb{E}[\int_0^T H^2 \, dt] < \infty$). The Ito isometry shows the integral map is an isometry, so it extends uniquely by continuity to all adapted, square-integrable processes.

### 2.2 Properties of the Ito Integral

Let $I(t) = \int_0^t H(s) \, dB(s)$ where $H$ is adapted and $\mathbb{E}[\int_0^T H^2 \, ds] < \infty$. Then:

1. **Martingale property:** $I(t)$ is a (continuous) martingale with respect to the filtration $\{\mathcal{F}_t\}$.

2. **Zero mean:** $\mathbb{E}[I(t)] = 0$ for all $t$.

3. **Ito isometry:**
$$
\mathbb{E}[I(t)^2] = \mathbb{E}\left[\int_0^t H(s)^2 \, ds\right]
$$

4. **Quadratic variation:**
$$
[I, I]_t = \int_0^t H(s)^2 \, ds
$$

5. **Linearity:** $\int_0^t (\alpha H + \beta K) \, dB = \alpha \int_0^t H \, dB + \beta \int_0^t K \, dB$.

6. **Not ordinary calculus:** $\int_0^t B(s) \, dB(s) = \frac{1}{2}B(t)^2 - \frac{1}{2}t \neq \frac{1}{2}B(t)^2$. The "missing" $-\frac{1}{2}t$ is the Ito correction.

> **Interview Tip:** The most common question about the Ito integral is: "Compute $\int_0^t B(s) \, dB(s)$." The answer is $\frac{1}{2}B(t)^2 - \frac{1}{2}t$. Be prepared to derive this from the definition or from Ito's lemma applied to $f(x) = x^2/2$. The extra $-t/2$ term is the hallmark of Ito calculus.

### 2.3 Ito vs Stratonovich

The Stratonovich integral uses the **midpoint** rule:

$$
\int_0^T H(t) \circ dB(t) = \lim \sum_i H\left(\frac{t_i + t_{i+1}}{2}\right)(B(t_{i+1}) - B(t_i))
$$

The relation between the two is:

$$
\int_0^T H(t) \circ dB(t) = \int_0^T H(t) \, dB(t) + \frac{1}{2}[H, B]_T
$$

For $H(t) = g(B(t))$, this becomes:

$$
\int_0^T g(B(t)) \circ dB(t) = \int_0^T g(B(t)) \, dB(t) + \frac{1}{2}\int_0^T g'(B(t)) \, dt
$$

Stratonovich calculus obeys ordinary chain rule, but the integral is **not** a martingale. In finance, we almost always use **Ito** because:
- The Ito integral is a martingale (essential for pricing).
- Non-anticipating (adapted) integrands model the fact that trading strategies cannot look into the future.

> **Interview Tip:** If asked "Why Ito and not Stratonovich?", explain that the Ito convention gives a martingale and reflects the economic reality that traders cannot anticipate future prices. The Stratonovich integral is useful in physics but not in finance.

## 3. Ito's Lemma

Ito's lemma is the chain rule of stochastic calculus -- the single most important computational tool in quantitative finance.

### 3.1 Statement

> **Definition (Ito process):** A process $X(t)$ is an Ito process if it can be written as:
>
> $$
> dX(t) = \mu(t) \, dt + \sigma(t) \, dB(t)
> $$
>
> or equivalently $X(t) = X(0) + \int_0^t \mu(s) \, ds + \int_0^t \sigma(s) \, dB(s)$.

**Ito's Lemma (one-dimensional).** Let $X(t)$ be an Ito process and $f(t, x) \in C^{1,2}$. Then $Y(t) = f(t, X(t))$ satisfies:

$$
df = \frac{\partial f}{\partial t} \, dt + \frac{\partial f}{\partial x} \, dX + \frac{1}{2} \frac{\partial^2 f}{\partial x^2} (dX)^2
$$

Expanding $(dX)^2 = \sigma^2 \, dt$ (using the multiplication table), we get the full form:

$$
\boxed{df = \left(\frac{\partial f}{\partial t} + \mu \frac{\partial f}{\partial x} + \frac{1}{2}\sigma^2 \frac{\partial^2 f}{\partial x^2}\right) dt + \sigma \frac{\partial f}{\partial x} \, dB}
$$

**Multi-dimensional version.** For $X = (X_1, \ldots, X_n)$ with $dX_i = \mu_i \, dt + \sum_j \sigma_{ij} \, dB_j$:

$$
df = \frac{\partial f}{\partial t} dt + \sum_i \frac{\partial f}{\partial x_i} dX_i + \frac{1}{2} \sum_{i,j} \frac{\partial^2 f}{\partial x_i \partial x_j} d[X_i, X_j]
$$

### 3.2 Proof Sketch

The proof proceeds by Taylor expanding $f(t + \Delta t, X + \Delta X)$ to second order:

$$
\Delta f \approx f_t \Delta t + f_x \Delta X + \frac{1}{2} f_{xx} (\Delta X)^2 + f_{tx} \Delta t \Delta X + \frac{1}{2} f_{tt} (\Delta t)^2
$$

Now substitute $\Delta X = \mu \Delta t + \sigma \Delta B$:

$$
(\Delta X)^2 = \sigma^2 (\Delta B)^2 + 2\mu\sigma \Delta t \Delta B + \mu^2 (\Delta t)^2
$$

Using the multiplication rules: $(\Delta B)^2 \to dt$ in the limit, while $\Delta t \cdot \Delta B$ and $(\Delta t)^2$ are $o(dt)$. Therefore only the $\frac{1}{2}\sigma^2 f_{xx} \, dt$ term survives from the second-order expansion. All higher-order terms vanish.

The rigorous proof sums over a partition, uses the quadratic variation result $\sum (\Delta B_i)^2 \to t$, and takes the mesh to zero.

### 3.3 Key Examples

#### Example 1: $d(B(t)^2)$

> **Example:** Let $f(x) = x^2$ and $X(t) = B(t)$ (so $\mu = 0$, $\sigma = 1$). Then $f' = 2x$, $f'' = 2$:
>
> $$
> d(B^2) = 2B \, dB + \frac{1}{2}(2)(1)^2 \, dt = 2B \, dB + dt
> $$
>
> Rearranging: $\int_0^t B \, dB = \frac{1}{2}(B(t)^2 - t)$.

#### Example 2: $d(e^{B(t)})$

> **Example:** Let $f(x) = e^x$ with $X = B$. Then $f' = f'' = e^x$:
>
> $$
> d(e^B) = e^B \, dB + \frac{1}{2} e^B \, dt = e^B\left(dB + \frac{1}{2}dt\right)
> $$
>
> In integral form: $e^{B(t)} = 1 + \int_0^t e^{B(s)} \, dB(s) + \frac{1}{2}\int_0^t e^{B(s)} \, ds$.

#### Example 3: $d(B(t)^2 - t)$

> **Example:** From Example 1, $d(B^2 - t) = 2B \, dB + dt - dt = 2B \, dB$.
>
> Since the $dt$ term vanishes, $B(t)^2 - t$ is a **martingale** (it is a pure stochastic integral with no drift).

#### Example 4: The Exponential Martingale $d(e^{\sigma B - \sigma^2 t/2})$

> **Example:** Let $f(t, x) = e^{\sigma x - \sigma^2 t/2}$ with $X = B$. Then:
>
> $$
> f_t = -\frac{\sigma^2}{2} f, \qquad f_x = \sigma f, \qquad f_{xx} = \sigma^2 f
> $$
>
> Applying Ito's lemma:
>
> $$
> df = \left(-\frac{\sigma^2}{2}f + 0 + \frac{1}{2}\sigma^2 f\right) dt + \sigma f \, dB = \sigma f \, dB
> $$
>
> The drift cancels exactly! So $\mathcal{E}(t) = e^{\sigma B(t) - \sigma^2 t/2}$ is a **martingale** (the stochastic or Doleans-Dade exponential). This is fundamental to Girsanov's theorem.

> **Interview Tip:** Memorize these four examples cold. They appear in interviews constantly. The exponential martingale $\mathcal{E}(t) = \exp(\sigma B - \sigma^2 t/2)$ is especially important: it is the Radon-Nikodym derivative in Girsanov's theorem and the backbone of risk-neutral pricing.

### 3.4 Product Rule (Ito Product Rule)

For two Ito processes $X$ and $Y$:

$$
d(XY) = X \, dY + Y \, dX + d[X, Y]
$$

where $[X, Y]_t$ is the cross-variation (covariation). If $dX = \mu_X dt + \sigma_X dB$ and $dY = \mu_Y dt + \sigma_Y dB$, then:

$$
d[X, Y] = \sigma_X \sigma_Y \, dt
$$

This extra $d[X,Y]$ term is the "Ito correction" for the product rule and is absent in ordinary calculus.

> **Example:** Verify: $d(B \cdot t) = B \, dt + t \, dB + d[B, t]$. Since $t$ has $\sigma = 0$, the cross-variation $d[B, t] = 0$, so $d(tB) = B \, dt + t \, dB$ -- same as ordinary calculus in this case.

## 4. Stochastic Differential Equations

A stochastic differential equation (SDE) takes the general form:

$$
dX(t) = \mu(t, X(t)) \, dt + \sigma(t, X(t)) \, dB(t), \qquad X(0) = x_0
$$

### 4.1 Geometric Brownian Motion (GBM)

The most important SDE in finance:

$$
dS = \mu S \, dt + \sigma S \, dB, \qquad S(0) = S_0
$$

**Solution.** Apply Ito's lemma to $f(S) = \ln S$:

$$
d(\ln S) = \frac{1}{S} dS - \frac{1}{2} \frac{1}{S^2} (dS)^2 = \frac{1}{S}(\mu S \, dt + \sigma S \, dB) - \frac{1}{2}\sigma^2 dt
$$

$$
= \left(\mu - \frac{\sigma^2}{2}\right) dt + \sigma \, dB
$$

Integrating from $0$ to $t$:

$$
\ln S(t) - \ln S_0 = \left(\mu - \frac{\sigma^2}{2}\right)t + \sigma B(t)
$$

$$
\boxed{S(t) = S_0 \exp\left[\left(\mu - \frac{\sigma^2}{2}\right)t + \sigma B(t)\right]}
$$

Key properties:
- $S(t)$ is **log-normally** distributed.
- $\mathbb{E}[S(t)] = S_0 e^{\mu t}$ (not $S_0 e^{(\mu - \sigma^2/2)t}$ -- the convexity correction matters).
- The $-\sigma^2/2$ term is the **Ito correction** and is the source of many interview questions.

> **Interview Tip:** You will almost certainly be asked to solve the GBM SDE. The key insight is to apply Ito's lemma to $\ln S$ and use $(dS)^2 = \sigma^2 S^2 \, dt$. Always remember that $\mathbb{E}[S(t)] = S_0 e^{\mu t}$, NOT $S_0 e^{(\mu - \sigma^2/2)t}$. The latter is the *median*, not the mean.

### 4.2 Ornstein-Uhlenbeck Process

$$
dX = \kappa(\theta - X) \, dt + \sigma \, dB, \qquad X(0) = x_0
$$

This is a **mean-reverting** process used for interest rates (Vasicek model), volatility, and pairs trading.

**Solution by integrating factor.** Multiply by $e^{\kappa t}$. Using the product rule:

$$
d(e^{\kappa t} X) = e^{\kappa t} dX + \kappa e^{\kappa t} X \, dt = e^{\kappa t}[\kappa(\theta - X) \, dt + \sigma \, dB] + \kappa e^{\kappa t} X \, dt
$$

$$
= \kappa \theta e^{\kappa t} \, dt + \sigma e^{\kappa t} \, dB
$$

Integrating:

$$
\boxed{X(t) = \theta + (x_0 - \theta)e^{-\kappa t} + \sigma \int_0^t e^{-\kappa(t-s)} \, dB(s)}
$$

Properties:
- $\mathbb{E}[X(t)] = \theta + (x_0 - \theta)e^{-\kappa t} \to \theta$ as $t \to \infty$.
- $\text{Var}(X(t)) = \frac{\sigma^2}{2\kappa}(1 - e^{-2\kappa t}) \to \frac{\sigma^2}{2\kappa}$ as $t \to \infty$.
- $X(t)$ is Gaussian (since it is a deterministic function plus a stochastic integral against BM).
- The stationary distribution is $\mathcal{N}\left(\theta, \frac{\sigma^2}{2\kappa}\right)$.

> **Interview Tip:** Be prepared to derive the OU solution using the integrating factor $e^{\kappa t}$. A common follow-up is: "What is the long-run distribution?" Answer: $\mathcal{N}(\theta, \sigma^2/(2\kappa))$.

### 4.3 Cox-Ingersoll-Ross (CIR) Process

$$
dX = \kappa(\theta - X) \, dt + \sigma \sqrt{X} \, dB, \qquad X(0) = x_0 > 0
$$

Key properties:
- If $2\kappa\theta \geq \sigma^2$ (the **Feller condition**), then $X(t) > 0$ for all $t$ a.s. This ensures rates/volatility stay positive.
- If $2\kappa\theta < \sigma^2$, the process can touch zero but is immediately reflected.
- The stationary distribution is a **Gamma distribution** with shape $2\kappa\theta/\sigma^2$ and rate $2\kappa/\sigma^2$.
- Unlike OU, CIR is **not Gaussian** (the $\sqrt{X}$ diffusion coefficient breaks Gaussianity).
- No closed-form solution exists; the transition density involves a non-central chi-squared distribution.

Applications: Heston stochastic volatility (for $v_t$), Cox-Ingersoll-Ross interest rate model.

> **Interview Tip:** The Feller condition $2\kappa\theta \geq \sigma^2$ is a classic interview question. Be ready to explain intuitively why it works: when $X$ is near zero, the mean-reversion drift $\kappa\theta$ pushes $X$ up, while the diffusion $\sigma\sqrt{X} \to 0$. If the drift is strong enough relative to the diffusion, $X$ cannot reach zero.

### 4.4 Strong vs Weak Solutions

> **Definition:** 
>
> - A **strong solution** of an SDE is a process $X(t)$ that is adapted to the filtration generated by the given Brownian motion $B(t)$, i.e., $X(t) = F(t, \{B(s) : 0 \leq s \leq t\})$ is a functional of the specific Brownian motion.
>
> - A **weak solution** is a probability space $(\Omega, \mathcal{F}, \mathbb{P})$, a filtration, a Brownian motion $B$, and a process $X$ satisfying the SDE. The Brownian motion may be constructed as part of the solution.

The distinction is subtle but important:

- **Strong existence + uniqueness** (pathwise): guaranteed under **Lipschitz conditions** on $\mu$ and $\sigma$ (the classical Ito existence/uniqueness theorem). GBM and OU both satisfy this.
- **Weak existence + uniqueness in law**: holds under much weaker conditions. For example, Tanaka's SDE $dX = \text{sgn}(X) \, dB$ has a weak solution but no strong solution.
- **Yamada-Watanabe theorem**: Weak existence + pathwise uniqueness $\Rightarrow$ strong existence.

| SDE | Strong Solution? | Notes |
|---|---|---|
| GBM: $dS = \mu S \, dt + \sigma S \, dB$ | Yes | Lipschitz, explicit formula |
| OU: $dX = \kappa(\theta - X)dt + \sigma \, dB$ | Yes | Linear, explicit formula |
| CIR: $dX = \kappa(\theta - X)dt + \sigma\sqrt{X} \, dB$ | Yes | Yamada-Watanabe applies |
| Tanaka: $dX = \text{sgn}(X) \, dB$ | No | Weak solution only |

## 5. Martingale Theory

Martingale theory is the mathematical foundation of fair pricing in finance.

### 5.1 Definition and Basic Properties

> **Definition:** A stochastic process $\{M(t)\}_{t \geq 0}$ adapted to a filtration $\{\mathcal{F}_t\}$ is a **martingale** if:
>
> 1. $\mathbb{E}[|M(t)|] < \infty$ for all $t$ (integrability).
> 2. $\mathbb{E}[M(t) \mid \mathcal{F}_s] = M(s)$ for all $s \leq t$ (the martingale property).
>
> If $\mathbb{E}[M(t) \mid \mathcal{F}_s] \leq M(s)$, then $M$ is a **supermartingale** (expected to decrease).
> If $\mathbb{E}[M(t) \mid \mathcal{F}_s] \geq M(s)$, then $M$ is a **submartingale** (expected to increase).

Intuition: a martingale is a "fair game" -- the best prediction of the future value, given current information, is the current value.

**Fundamental consequence:** If $M$ is a martingale, then $\mathbb{E}[M(t)] = \mathbb{E}[M(0)]$ for all $t$.

### 5.2 Important Examples of Martingales

| Process | Martingale? | Proof sketch |
|---|---|---|
| $B(t)$ | Yes | $\mathbb{E}[B(t) \mid \mathcal{F}_s] = B(s) + \mathbb{E}[B(t)-B(s) \mid \mathcal{F}_s] = B(s)$ |
| $B(t)^2 - t$ | Yes | Ito's lemma: $d(B^2 - t) = 2B \, dB$ (no drift) |
| $B(t)^3 - 3tB(t)$ | Yes | Ito's lemma on $f(t,x) = x^3 - 3tx$ |
| $\exp(\sigma B(t) - \sigma^2 t/2)$ | Yes | Ito's lemma: drift cancels (shown in Section 3.3) |
| $\int_0^t H(s) \, dB(s)$ | Yes | Ito integral with $\mathbb{E}[\int H^2 \, ds] < \infty$ |
| $B(t)^2$ | No | $\mathbb{E}[B(t)^2] = t \neq s = \mathbb{E}[B(s)^2]$ for $t \neq s$ |
| $e^{B(t)}$ | No | $\mathbb{E}[e^{B(t)}] = e^{t/2}$ is increasing |

**Verification technique:** Apply Ito's lemma to $f(t, B(t))$. If the $dt$ drift term vanishes, then $f(t, B(t))$ is a local martingale. Under integrability conditions, it is a true martingale.

> **Example:** Verify $M(t) = B(t)^3 - 3tB(t)$ is a martingale.
>
> Let $f(t,x) = x^3 - 3tx$. Then $f_t = -3x$, $f_x = 3x^2 - 3t$, $f_{xx} = 6x$.
>
> $$
> df = (-3B + 0 + \frac{1}{2}(6B)) \, dt + (3B^2 - 3t) \, dB = 0 \cdot dt + (3B^2 - 3t) \, dB
> $$
>
> The drift is zero, so $M(t)$ is a (local) martingale.

### 5.3 Compensated Poisson Process

Let $N(t)$ be a Poisson process with intensity $\lambda$. The **compensated Poisson process**:

$$
\tilde{N}(t) = N(t) - \lambda t
$$

is a martingale. This follows because:

$$
\mathbb{E}[\tilde{N}(t) \mid \mathcal{F}_s] = \mathbb{E}[N(t) - \lambda t \mid \mathcal{F}_s] = N(s) + \lambda(t-s) - \lambda t = N(s) - \lambda s = \tilde{N}(s)
$$

More generally, $\exp(\theta N(t) - \lambda t (e^\theta - 1))$ is a martingale for any $\theta$ (the exponential martingale for Poisson processes).

This is important in **jump-diffusion models** (Merton's model) where asset prices can have both continuous and jump components.

### 5.4 Optional Stopping Theorem

**Theorem (Optional Stopping / Optional Sampling).** Let $M(t)$ be a continuous martingale and $\tau$ a stopping time. If any of the following sufficient conditions hold:

1. $\tau$ is bounded: $\tau \leq T$ a.s. for some constant $T$, or
2. $M$ is uniformly integrable and $\tau < \infty$ a.s., or
3. $\mathbb{E}[\tau] < \infty$ and there exists $c$ such that $|M(t \wedge \tau) - M((t-1)^+ \wedge \tau)| \leq c$ a.s.

then $\mathbb{E}[M(\tau)] = \mathbb{E}[M(0)]$.

**Warning:** Without these conditions, the theorem can fail spectacularly.

> **Example (Classic gambler's ruin):** Let $B(t)$ be BM and $\tau = \inf\{t : B(t) \notin (-a, b)\}$ for $a, b > 0$. Since $B$ is a martingale, by optional stopping:
>
> $$
> 0 = \mathbb{E}[B(0)] = \mathbb{E}[B(\tau)] = (-a) \cdot \mathbb{P}(B(\tau) = -a) + b \cdot \mathbb{P}(B(\tau) = b)
> $$
>
> Combined with $\mathbb{P}(B(\tau) = -a) + \mathbb{P}(B(\tau) = b) = 1$, we get:
>
> $$
> \mathbb{P}(B(\tau) = b) = \frac{a}{a+b}, \qquad \mathbb{P}(B(\tau) = -a) = \frac{b}{a+b}
> $$
>
> Applying optional stopping to the martingale $B(t)^2 - t$ gives $\mathbb{E}[\tau] = ab$.

> **Interview Tip:** Optional stopping theorem problems are interview favorites. The standard approach is: (1) Identify the martingale ($B$, $B^2 - t$, or $e^{\theta B - \theta^2 t/2}$). (2) Verify the conditions for OST. (3) Apply $\mathbb{E}[M(\tau)] = \mathbb{E}[M(0)]$. Using multiple martingales gives you a system of equations.

### 5.5 Doob's Maximal Inequality

**Theorem (Doob's $L^p$ Maximal Inequality).** Let $M(t)$ be a non-negative submartingale (or the absolute value of a martingale). Then for $p > 1$:

$$
\mathbb{E}\left[\sup_{0 \leq s \leq t} |M(s)|^p\right] \leq \left(\frac{p}{p-1}\right)^p \mathbb{E}[|M(t)|^p]
$$

For $p = 1$ (Doob's maximal inequality):

$$
\mathbb{P}\left(\sup_{0 \leq s \leq t} M(s) \geq \lambda\right) \leq \frac{\mathbb{E}[M(t)]}{\lambda}
$$

**Application:** For Brownian motion, with $p = 2$:

$$
\mathbb{E}\left[\sup_{0 \leq s \leq t} B(s)^2\right] \leq 4 \, \mathbb{E}[B(t)^2] = 4t
$$

This is crucial in proving convergence results and bounding the behavior of stochastic processes.

### 5.6 Martingale Representation Theorem

**Theorem.** Let $M(t)$ be a square-integrable martingale with respect to the filtration generated by a Brownian motion $B(t)$. Then there exists a unique adapted process $H(t)$ with $\mathbb{E}[\int_0^T H^2 \, ds] < \infty$ such that:

$$
M(t) = M(0) + \int_0^t H(s) \, dB(s)
$$

**Financial interpretation:** Every contingent claim (martingale under the risk-neutral measure) can be replicated by a trading strategy $H(t)$ (the hedge ratio). This is the mathematical foundation of **complete markets** and **hedging**.

> **Interview Tip:** The martingale representation theorem guarantees that in the Black-Scholes model, every derivative can be perfectly hedged. If asked "Why is the Black-Scholes market complete?", the answer traces back to this theorem.

## 6. Girsanov's Theorem

Girsanov's theorem is the mathematical engine behind risk-neutral pricing: it allows us to change the probability measure so that discounted asset prices become martingales.

### 6.1 Statement

**Theorem (Girsanov).** Let $B(t)$ be a standard Brownian motion under measure $\mathbb{P}$, and let $\gamma(t)$ be an adapted process satisfying the **Novikov condition**:

$$
\mathbb{E}^{\mathbb{P}}\left[\exp\left(\frac{1}{2}\int_0^T \gamma(s)^2 \, ds\right)\right] < \infty
$$

Define the **Radon-Nikodym derivative** (likelihood ratio process):

$$
Z(t) = \frac{d\mathbb{Q}}{d\mathbb{P}}\bigg|_{\mathcal{F}_t} = \exp\left(-\int_0^t \gamma(s) \, dB(s) - \frac{1}{2}\int_0^t \gamma(s)^2 \, ds\right)
$$

Then under the new measure $\mathbb{Q}$, the process:

$$
\tilde{B}(t) = B(t) + \int_0^t \gamma(s) \, ds
$$

is a standard Brownian motion.

Equivalently, $B(t) = \tilde{B}(t) - \int_0^t \gamma(s) \, ds$, so under $\mathbb{Q}$:

$$
dB(t) = d\tilde{B}(t) - \gamma(t) \, dt
$$

Girsanov's theorem **removes drift** by changing the measure.

### 6.2 Connection to Risk-Neutral Pricing

Consider a stock following GBM under the real-world measure $\mathbb{P}$:

$$
dS = \mu S \, dt + \sigma S \, dB
$$

The discounted price $\tilde{S}(t) = e^{-rt}S(t)$ has dynamics:

$$
d\tilde{S} = (\mu - r)\tilde{S} \, dt + \sigma \tilde{S} \, dB
$$

For $\tilde{S}$ to be a martingale, the drift must vanish. Set $\gamma = \frac{\mu - r}{\sigma}$ (the **market price of risk** or **Sharpe ratio**). By Girsanov:

$$
\tilde{B}(t) = B(t) + \frac{\mu - r}{\sigma} t
$$

is a BM under $\mathbb{Q}$. Substituting $dB = d\tilde{B} - \frac{\mu - r}{\sigma} dt$:

$$
dS = \mu S \, dt + \sigma S\left(d\tilde{B} - \frac{\mu - r}{\sigma} dt\right) = rS \, dt + \sigma S \, d\tilde{B}
$$

Under $\mathbb{Q}$, the stock grows at the **risk-free rate** $r$, and $\tilde{S}(t) = e^{-rt}S(t)$ is a $\mathbb{Q}$-martingale.

The **fundamental pricing formula** is:

$$
\boxed{V(0) = e^{-rT} \, \mathbb{E}^{\mathbb{Q}}[\text{Payoff}(S(T))]}
$$

> **Interview Tip:** When asked about risk-neutral pricing, explain the three steps: (1) Identify the market price of risk $\gamma = (\mu - r)/\sigma$. (2) Apply Girsanov to get a new BM $\tilde{B}$ under $\mathbb{Q}$. (3) Under $\mathbb{Q}$, the stock drift becomes $r$ and discounted prices are martingales. The actual drift $\mu$ is irrelevant for pricing -- this is perhaps the deepest insight in mathematical finance.

### 6.3 The Radon-Nikodym Derivative in Detail

The process $Z(t) = \mathcal{E}(-\gamma \cdot B)_t$ is the exponential martingale:

$$
Z(t) = \exp\left(-\gamma B(t) - \frac{1}{2}\gamma^2 t\right)
$$

(for constant $\gamma$). By Ito's lemma, $dZ = -\gamma Z \, dB$, confirming $Z$ is a $\mathbb{P}$-martingale with $\mathbb{E}^{\mathbb{P}}[Z(t)] = 1$.

The change-of-measure formula for expectations is:

$$
\mathbb{E}^{\mathbb{Q}}[X] = \mathbb{E}^{\mathbb{P}}[Z(T) \cdot X]
$$

This is the **abstract Bayes formula** and is the computational workhorse when switching between $\mathbb{P}$ and $\mathbb{Q}$.

> **Interview Tip:** A sharp interviewer might ask: "What conditions ensure $Z(t)$ is a true martingale and not just a local martingale?" The answer is the **Novikov condition**. If Novikov fails, $Z$ could be a strict local martingale with $\mathbb{E}[Z(T)] < 1$, and the change of measure breaks down. This is related to financial bubbles in some models.

## 7. Feynman-Kac Formula

The Feynman-Kac formula provides a profound link between partial differential equations (PDEs) and stochastic processes.

### 7.1 Statement

**Theorem (Feynman-Kac).** Consider the PDE:

$$
\frac{\partial u}{\partial t} + \mu(x)\frac{\partial u}{\partial x} + \frac{1}{2}\sigma(x)^2 \frac{\partial^2 u}{\partial x^2} - r(x) u = 0
$$

with terminal condition $u(T, x) = \phi(x)$. Then the solution is given by the **stochastic representation**:

$$
\boxed{u(t, x) = \mathbb{E}\left[e^{-\int_t^T r(X(s)) \, ds} \phi(X(T)) \,\bigg|\, X(t) = x\right]}
$$

where $X$ satisfies the SDE $dX = \mu(X) \, dt + \sigma(X) \, dB$.

**Special case** ($r$ constant, $\mu(x) = rx$, $\sigma(x) = \sigma x$): This gives exactly the Black-Scholes PDE, and the Feynman-Kac solution is:

$$
u(t, x) = e^{-r(T-t)} \mathbb{E}^{\mathbb{Q}}[\phi(S(T)) \mid S(t) = x]
$$

which is precisely the risk-neutral pricing formula.

### 7.2 Proof Sketch

Define $Y(s) = e^{-\int_t^s r(X(u)) \, du} u(s, X(s))$ for $s \in [t, T]$. Apply Ito's lemma to $Y$:

$$
dY = e^{-\int_t^s r \, du}\left[-r \cdot u + u_s + \mu u_x + \frac{1}{2}\sigma^2 u_{xx}\right] ds + e^{-\int_t^s r \, du} \sigma u_x \, dB
$$

The coefficient of $ds$ is zero by the PDE. So $Y(s)$ is a local martingale. Under suitable growth conditions, $Y$ is a true martingale, so:

$$
u(t, x) = Y(t) = \mathbb{E}[Y(T) \mid X(t) = x] = \mathbb{E}\left[e^{-\int_t^T r \, ds} \phi(X(T)) \mid X(t) = x\right]
$$

> **Interview Tip:** The Feynman-Kac formula connects the PDE approach (Black-Scholes equation) and the probabilistic approach (risk-neutral expectation) to derivatives pricing. Be ready to state this connection: solving the BS PDE with terminal condition $\phi(S)$ is the same as computing $e^{-r(T-t)}\mathbb{E}^\mathbb{Q}[\phi(S_T)]$.

### 7.3 Applications and Extensions

**1. Kolmogorov backward equation.** Setting $r = 0$ and $\phi(x) = \mathbf{1}_{A}(x)$ gives:

$$
u(t,x) = \mathbb{P}(X(T) \in A \mid X(t) = x)
$$

which satisfies $u_t + \mu u_x + \frac{1}{2}\sigma^2 u_{xx} = 0$.

**2. Bond pricing.** In the Vasicek model ($dr = \kappa(\theta - r)dt + \sigma dB$), the zero-coupon bond price is:

$$
P(t, T) = \mathbb{E}^\mathbb{Q}\left[e^{-\int_t^T r(s) \, ds} \,\bigg|\, r(t)\right]
$$

By Feynman-Kac, this satisfies a PDE in $(t, r)$ space.

**3. Killed diffusion.** With $r(x) > 0$, the Feynman-Kac formula accounts for "killing" (discounting). This arises naturally in barrier option pricing where the process is killed upon hitting a barrier.

## 8. Applications: Black-Scholes

The Black-Scholes model is the crown jewel of mathematical finance, bringing together all the stochastic calculus machinery developed above.

### 8.1 Derivation via Ito's Lemma (PDE Approach)

**Setup:** Stock price follows GBM: $dS = \mu S \, dt + \sigma S \, dB$. A derivative has price $V(t, S)$.

**Step 1: Apply Ito's lemma to $V(t, S)$:**

$$
dV = \left(V_t + \mu S V_S + \frac{1}{2}\sigma^2 S^2 V_{SS}\right) dt + \sigma S V_S \, dB
$$

**Step 2: Construct the replicating (hedging) portfolio.** Hold $\Delta = V_S$ shares of stock and $-1$ derivative. The portfolio value is $\Pi = \Delta S - V$, and:

$$
d\Pi = \Delta \, dS - dV = V_S \, dS - dV
$$

$$
= V_S(\mu S \, dt + \sigma S \, dB) - \left(V_t + \mu S V_S + \frac{1}{2}\sigma^2 S^2 V_{SS}\right) dt - \sigma S V_S \, dB
$$

$$
= -\left(V_t + \frac{1}{2}\sigma^2 S^2 V_{SS}\right) dt
$$

The $dB$ terms cancel -- the portfolio is **riskless**!

**Step 3: No-arbitrage argument.** A riskless portfolio must earn the risk-free rate:

$$
d\Pi = r \Pi \, dt = r(V_S S - V) \, dt
$$

Setting the two expressions for $d\Pi$ equal:

$$
-V_t - \frac{1}{2}\sigma^2 S^2 V_{SS} = r(V_S S - V)
$$

$$
\boxed{V_t + rS V_S + \frac{1}{2}\sigma^2 S^2 V_{SS} = rV}
$$

This is the **Black-Scholes PDE**. Notice that $\mu$ does not appear -- the derivative price is independent of the stock's expected return!

> **Interview Tip:** The disappearance of $\mu$ from the BS PDE is one of the most profound results in finance. If asked "Why doesn't $\mu$ appear?", explain that the hedging argument eliminates the randomness, and by no-arbitrage, the resulting riskless portfolio must grow at rate $r$. The drift $\mu$ is absorbed into the market price of risk via Girsanov's theorem.

### 8.2 Risk-Neutral Valuation Approach

Under $\mathbb{Q}$, $dS = rS \, dt + \sigma S \, d\tilde{B}$, so:

$$
S(T) = S(0) \exp\left[\left(r - \frac{\sigma^2}{2}\right)T + \sigma \tilde{B}(T)\right]
$$

For a European call with payoff $(S(T) - K)^+$:

$$
C = e^{-rT} \mathbb{E}^{\mathbb{Q}}[(S(T) - K)^+]
$$

Let $Z = \tilde{B}(T)/\sqrt{T} \sim \mathcal{N}(0,1)$ under $\mathbb{Q}$. Then $S(T) = S_0 e^{(r - \sigma^2/2)T + \sigma\sqrt{T}Z}$.

$S(T) > K$ when $Z > -d_2$ where:

$$
d_2 = \frac{\ln(S_0/K) + (r - \sigma^2/2)T}{\sigma\sqrt{T}}, \qquad d_1 = d_2 + \sigma\sqrt{T}
$$

Computing the expectation (splitting into $\mathbb{E}[S(T) \mathbf{1}_{S(T)>K}]$ and $K \cdot \mathbb{P}(S(T) > K)$) gives:

$$
\boxed{C = S_0 \Phi(d_1) - K e^{-rT} \Phi(d_2)}
$$

$$
\boxed{P = K e^{-rT} \Phi(-d_2) - S_0 \Phi(-d_1)}
$$

where $\Phi$ is the standard normal CDF.

### 8.3 The Greeks via Stochastic Calculus

The Greeks are the sensitivities of the option price. They arise naturally from the Ito expansion.

From the BS PDE $V_t + rSV_S + \frac{1}{2}\sigma^2 S^2 V_{SS} = rV$:

| Greek | Definition | Role in BS PDE |
|---|---|---|
| $\Delta = V_S$ | Sensitivity to $S$ | Hedge ratio, coefficient of $dS$ in Ito expansion |
| $\Gamma = V_{SS}$ | Sensitivity of Delta to $S$ | Source of the Ito correction term |
| $\Theta = V_t$ | Sensitivity to time | Time decay |
| Vega $= V_\sigma$ | Sensitivity to $\sigma$ | Not in BS PDE directly (constant $\sigma$) |
| $\rho = V_r$ | Sensitivity to $r$ | Small effect for short-dated options |

The BS PDE can be rewritten as the **theta-gamma relation**:

$$
\Theta + rS\Delta + \frac{1}{2}\sigma^2 S^2 \Gamma = rV
$$

This is the fundamental P&L equation: the daily P&L of a delta-hedged portfolio is:

$$
\text{P\&L} \approx \Theta \, \Delta t + \frac{1}{2}\Gamma (\Delta S)^2
$$

If $\Gamma > 0$ (long options), you profit from large moves but pay theta. If $\Gamma < 0$, the reverse.

> **Interview Tip:** The theta-gamma tradeoff is a staple interview topic. A long gamma position benefits from realized volatility exceeding implied volatility, while theta provides the cost of maintaining the position. The break-even daily move is $\sqrt{2|\Theta|/\Gamma}$.

### 8.4 Hedging Argument and Market Completeness

The delta hedge $\Delta = V_S = \Phi(d_1)$ for a call option replicates the payoff:

$$
V(t, S(t)) = \Delta(t) \cdot S(t) + \text{Cash}(t)
$$

The self-financing condition ensures the replicating portfolio requires no additional capital injection. This works because:

1. **Martingale representation theorem** guarantees the existence of the hedge.
2. **Ito's lemma** provides the hedge ratio $\Delta = V_S$.
3. **Girsanov's theorem** ensures discounted prices are martingales under $\mathbb{Q}$.

**Market completeness** in the BS model means every contingent claim can be replicated. This relies on:
- One source of randomness (one BM) matched by one tradeable asset.
- Continuous trading.
- No transaction costs, no constraints.

When any of these fail (stochastic volatility, jumps, transaction costs), the market becomes **incomplete** and perfect hedging is impossible.

> **Interview Tip:** If asked about the limitations of Black-Scholes, key points are: (1) Constant volatility (violated by the volatility smile). (2) Continuous trading (impossible in practice). (3) No jumps (crashes happen). (4) Log-normal returns (real returns have fat tails). Understanding these limitations and how models like Heston, SABR, or local volatility address them is what separates a strong candidate.

## 9. Worked Interview Problems

Below are detailed solutions to classic stochastic calculus interview problems.

### Problem 1: Compute $\mathbb{E}[B(t)^3]$ and $\mathbb{E}[B(t)^4]$

**Part (a): $\mathbb{E}[B(t)^3]$**

Since $B(t) \sim \mathcal{N}(0, t)$, by the symmetry of the normal distribution, all odd moments are zero:

$$
\boxed{\mathbb{E}[B(t)^3] = 0}
$$

**Alternative via Ito's lemma.** Apply Ito to $f(x) = x^3$ with $X = B$:

$$
d(B^3) = 3B^2 \, dB + \frac{1}{2}(6B) \, dt = 3B^2 \, dB + 3B \, dt
$$

Integrating: $B(t)^3 = 3\int_0^t B^2 \, dB + 3\int_0^t B \, ds$.

Taking expectations (the Ito integral has zero expectation):

$$
\mathbb{E}[B(t)^3] = 3\int_0^t \mathbb{E}[B(s)] \, ds = 3 \int_0^t 0 \, ds = 0 \quad \checkmark
$$

---

**Part (b): $\mathbb{E}[B(t)^4]$**

**Method 1: Moment generating function.** For $Z \sim \mathcal{N}(0, 1)$, $B(t) = \sqrt{t} \, Z$, so:

$$
\mathbb{E}[B(t)^4] = t^2 \mathbb{E}[Z^4] = t^2 \cdot 3 = 3t^2
$$

(using the fact that $\mathbb{E}[Z^{2n}] = (2n-1)!! = 1 \cdot 3 \cdot 5 \cdots (2n-1)$, so $\mathbb{E}[Z^4] = 3$).

**Method 2: Ito's lemma.** Apply Ito to $f(x) = x^4$:

$$
d(B^4) = 4B^3 \, dB + \frac{1}{2}(12B^2) \, dt = 4B^3 \, dB + 6B^2 \, dt
$$

Taking expectations:

$$
\mathbb{E}[B(t)^4] = 6\int_0^t \mathbb{E}[B(s)^2] \, ds = 6\int_0^t s \, ds = 6 \cdot \frac{t^2}{2} = 3t^2
$$

$$
\boxed{\mathbb{E}[B(t)^4] = 3t^2}
$$

**Method 3: Ito isometry (for $\mathbb{E}[B(t)^4]$).** Since $B(t) = \int_0^t dB$:

$$
B(t)^2 = 2\int_0^t B(s) \, dB(s) + t
$$

$$
\mathbb{E}[B(t)^4] = \mathbb{E}\left[\left(2\int_0^t B \, dB + t\right)^2\right] = 4\mathbb{E}\left[\left(\int_0^t B \, dB\right)^2\right] + 4t \cdot \underbrace{\mathbb{E}\left[\int_0^t B \, dB\right]}_{=0} + t^2
$$

By Ito isometry: $\mathbb{E}\left[\left(\int_0^t B \, dB\right)^2\right] = \int_0^t \mathbb{E}[B(s)^2] \, ds = t^2/2$.

So $\mathbb{E}[B(t)^4] = 4 \cdot t^2/2 + t^2 = 3t^2$. $\checkmark$

> **Interview Tip:** Showing multiple methods demonstrates deep understanding. The Ito's lemma method generalizes to compute any $\mathbb{E}[B(t)^n]$ recursively. The recursion is: $\mathbb{E}[B(t)^n] = \frac{n(n-1)}{2}\int_0^t \mathbb{E}[B(s)^{n-2}] \, ds$.

### Problem 2: Solve $dX = \mu X \, dt + \sigma X \, dB$ (GBM)

**Problem:** Find the explicit solution to geometric Brownian motion with $X(0) = x_0$.

**Solution:**

**Step 1:** Let $Y(t) = \ln X(t)$. Apply Ito's lemma with $f(x) = \ln x$, $f'(x) = 1/x$, $f''(x) = -1/x^2$:

$$
dY = \frac{1}{X} dX - \frac{1}{2} \frac{1}{X^2} (dX)^2
$$

**Step 2:** Compute $(dX)^2$:

$$
(dX)^2 = (\mu X \, dt + \sigma X \, dB)^2 = \sigma^2 X^2 \, dt
$$

(using $(dt)^2 = 0$, $dt \cdot dB = 0$, $(dB)^2 = dt$).

**Step 3:** Substitute:

$$
dY = \frac{\mu X \, dt + \sigma X \, dB}{X} - \frac{\sigma^2 X^2 \, dt}{2X^2} = \left(\mu - \frac{\sigma^2}{2}\right) dt + \sigma \, dB
$$

**Step 4:** Integrate from $0$ to $t$:

$$
Y(t) - Y(0) = \left(\mu - \frac{\sigma^2}{2}\right) t + \sigma B(t)
$$

**Step 5:** Exponentiate:

$$
\boxed{X(t) = x_0 \exp\left[\left(\mu - \frac{\sigma^2}{2}\right) t + \sigma B(t)\right]}
$$

**Verification.** Apply Ito's lemma to $g(t) = x_0 \exp\left[(\mu - \sigma^2/2)t + \sigma B(t)\right]$:

$$
dg = g \cdot \left[(\mu - \sigma^2/2) \, dt + \sigma \, dB\right] + \frac{1}{2} g \cdot \sigma^2 \, dt = g(\mu \, dt + \sigma \, dB) = \mu g \, dt + \sigma g \, dB
$$

which matches the original SDE. $\checkmark$

> **Interview Tip:** Always verify your SDE solution by plugging back in with Ito's lemma. This is a quick sanity check and demonstrates rigor. The $-\sigma^2/2$ Ito correction in the exponent is where many candidates make errors.

### Problem 3: Compute $\mathbb{E}[e^{B(t)}]$

**Problem:** Find $\mathbb{E}[e^{B(t)}]$ where $B(t)$ is standard Brownian motion.

**Method 1: Direct computation.**

Since $B(t) \sim \mathcal{N}(0, t)$, this is the moment generating function of a normal random variable evaluated at $\theta = 1$:

$$
\mathbb{E}[e^{\theta B(t)}] = \mathbb{E}[e^{\theta \sqrt{t} Z}] \quad \text{where } Z \sim \mathcal{N}(0,1)
$$

$$
= \int_{-\infty}^{\infty} e^{\theta \sqrt{t} z} \cdot \frac{1}{\sqrt{2\pi}} e^{-z^2/2} \, dz = \int_{-\infty}^{\infty} \frac{1}{\sqrt{2\pi}} e^{-(z^2 - 2\theta\sqrt{t}z)/2} \, dz
$$

Complete the square: $z^2 - 2\theta\sqrt{t}z = (z - \theta\sqrt{t})^2 - \theta^2 t$.

$$
= e^{\theta^2 t/2} \int_{-\infty}^{\infty} \frac{1}{\sqrt{2\pi}} e^{-(z - \theta\sqrt{t})^2/2} \, dz = e^{\theta^2 t/2}
$$

Setting $\theta = 1$:

$$
\boxed{\mathbb{E}[e^{B(t)}] = e^{t/2}}
$$

**Method 2: Via the exponential martingale.**

We know $M(t) = e^{\sigma B(t) - \sigma^2 t/2}$ is a martingale, so $\mathbb{E}[M(t)] = M(0) = 1$:

$$
\mathbb{E}[e^{\sigma B(t) - \sigma^2 t/2}] = 1 \implies \mathbb{E}[e^{\sigma B(t)}] = e^{\sigma^2 t/2}
$$

Setting $\sigma = 1$ gives $\mathbb{E}[e^{B(t)}] = e^{t/2}$. $\checkmark$

**Method 3: Ito's lemma + ODE.**

Let $m(t) = \mathbb{E}[e^{B(t)}]$. From the Ito expansion $d(e^B) = e^B \, dB + \frac{1}{2}e^B \, dt$, taking expectations:

$$
\frac{dm}{dt} = \frac{1}{2}m(t), \qquad m(0) = 1
$$

This ODE has solution $m(t) = e^{t/2}$. $\checkmark$

**More general:** $\mathbb{E}[e^{\alpha B(t) + \beta t}] = e^{(\alpha^2/2 + \beta)t}$.

> **Interview Tip:** Method 2 (using the exponential martingale) is the most elegant and quickest. It also shows deep understanding. Always mention that $e^{B(t)}$ is NOT a martingale because $\mathbb{E}[e^{B(t)}] = e^{t/2} \neq 1$. The correct martingale is $e^{B(t) - t/2}$.

### Problem 4: Stopping Time Problem

**Problem:** Let $B(t)$ be a standard Brownian motion. Define $\tau_a = \inf\{t \geq 0 : B(t) = a\}$ for $a > 0$. Find:

(a) $\mathbb{P}(\tau_a < \infty)$

(b) $\mathbb{E}[\tau_a]$

(c) $\mathbb{E}[e^{-\lambda \tau_a}]$ for $\lambda > 0$ (the Laplace transform)

---

**Solution (a):** By the law of the iterated logarithm (or by recurrence of 1-d BM):

$$
\limsup_{t \to \infty} B(t) = +\infty \quad \text{a.s.}
$$

Therefore $\mathbb{P}(\tau_a < \infty) = 1$ for all $a > 0$ (and by symmetry, for $a < 0$ as well).

---

**Solution (b):** Apply optional stopping to $B(t)^2 - t$ (a martingale) at $\tau_a \wedge T$:

$$
\mathbb{E}[B(\tau_a \wedge T)^2 - (\tau_a \wedge T)] = 0
$$

Since $|B(\tau_a \wedge T)| \leq a + \max_{s \leq T} |B(s) - B(\tau_a \wedge T)| $, and $B(\tau_a \wedge T)^2 \leq a^2$ on $\{\tau_a \leq T\}$ while $B(\tau_a \wedge T)^2 = B(T)^2$ on $\{\tau_a > T\}$:

$$
\mathbb{E}[\tau_a \wedge T] = \mathbb{E}[B(\tau_a \wedge T)^2] \leq \mathbb{E}[\max(a^2, B(T)^2)]
$$

Letting $T \to \infty$ with monotone convergence on the left side and dominated convergence analysis:

$$
\boxed{\mathbb{E}[\tau_a] = \infty}
$$

Even though BM hits level $a$ with probability 1, it takes infinite expected time! This can also be seen from the Laplace transform in part (c).

---

**Solution (c):** Use the exponential martingale $M(t) = e^{\theta B(t) - \theta^2 t/2}$.

Apply optional stopping at $\tau_a$ (the stopped martingale is bounded since $B(\tau_a) = a$):

$$
1 = \mathbb{E}[M(0)] = \mathbb{E}[M(\tau_a)] = \mathbb{E}[e^{\theta a - \theta^2 \tau_a/2}] = e^{\theta a} \mathbb{E}[e^{-\theta^2 \tau_a/2}]
$$

Set $\lambda = \theta^2/2$, i.e., $\theta = \sqrt{2\lambda}$:

$$
\boxed{\mathbb{E}[e^{-\lambda \tau_a}] = e^{-a\sqrt{2\lambda}}}
$$

Note that as $\lambda \to 0^+$, $\mathbb{E}[e^{-\lambda\tau_a}] \to 1$, confirming $\tau_a < \infty$ a.s. But $-\frac{d}{d\lambda}\mathbb{E}[e^{-\lambda\tau_a}]\big|_{\lambda=0} = \mathbb{E}[\tau_a]$ diverges, confirming $\mathbb{E}[\tau_a] = \infty$.

**Bonus:** The density of $\tau_a$ is the **inverse Gaussian** (also called the Levy distribution for this case):

$$
f_{\tau_a}(t) = \frac{a}{\sqrt{2\pi t^3}} \exp\left(-\frac{a^2}{2t}\right), \qquad t > 0
$$

> **Interview Tip:** This problem tests three key skills: (1) Identifying the right martingale to use. (2) Applying optional stopping correctly. (3) Interpreting the result ($\tau_a < \infty$ a.s. but $\mathbb{E}[\tau_a] = \infty$). The Laplace transform technique using the exponential martingale is extremely versatile for stopping time problems.

### Problem 5: Ito's Lemma Applications

#### Problem 5a: Show $B(t)^2 - t$ is a martingale

**Solution:** Apply Ito's lemma to $f(x) = x^2$ with $dX = dB$ ($\mu = 0$, $\sigma = 1$):

$$
d(B^2) = 2B \, dB + \frac{1}{2}(2)(1) \, dt = 2B \, dB + dt
$$

Therefore:

$$
d(B^2 - t) = 2B \, dB + dt - dt = 2B \, dB
$$

Since $B(t)^2 - t$ is a pure Ito integral (no $dt$ drift term) and $\mathbb{E}[\int_0^T 4B(s)^2 \, ds] = 4\int_0^T s \, ds = 2T^2 < \infty$, it is a (true) martingale. $\square$

---

#### Problem 5b: Show $\mathcal{E}(t) = e^{\sigma B(t) - \sigma^2 t/2}$ is a martingale and compute $d\mathcal{E}$

**Solution:** Let $f(t, x) = e^{\sigma x - \sigma^2 t / 2}$. Compute the partial derivatives:

$$
f_t = -\frac{\sigma^2}{2}f, \qquad f_x = \sigma f, \qquad f_{xx} = \sigma^2 f
$$

By Ito's lemma ($\mu = 0$, $\sigma_{\text{diff}} = 1$ for $dB$):

$$
d\mathcal{E} = f_t \, dt + f_x \, dB + \frac{1}{2}f_{xx} \, dt
$$

$$
= \left(-\frac{\sigma^2}{2}\mathcal{E} + \frac{1}{2}\sigma^2 \mathcal{E}\right) dt + \sigma \mathcal{E} \, dB
$$

$$
\boxed{d\mathcal{E} = \sigma \mathcal{E} \, dB}
$$

The drift term is exactly zero, confirming $\mathcal{E}$ is a local martingale. Since the Novikov condition holds ($\int_0^T \sigma^2 \, ds = \sigma^2 T < \infty$), $\mathcal{E}$ is a true martingale.

**Verification:** $\mathbb{E}[\mathcal{E}(t)] = \mathbb{E}[e^{\sigma B(t)}] \cdot e^{-\sigma^2 t/2} = e^{\sigma^2 t/2} \cdot e^{-\sigma^2 t/2} = 1 = \mathcal{E}(0)$. $\checkmark$

---

#### Problem 5c: Compute $d\left(\frac{1}{S(t)}\right)$ where $dS = \mu S \, dt + \sigma S \, dB$

**Solution:** Let $f(x) = 1/x$. Then $f'(x) = -1/x^2$, $f''(x) = 2/x^3$.

$$
d\left(\frac{1}{S}\right) = -\frac{1}{S^2} dS + \frac{1}{2} \cdot \frac{2}{S^3} (dS)^2
$$

$$
= -\frac{1}{S^2}(\mu S \, dt + \sigma S \, dB) + \frac{1}{S^3} \sigma^2 S^2 \, dt
$$

$$
= \left(-\frac{\mu}{S} + \frac{\sigma^2}{S}\right) dt - \frac{\sigma}{S} dB
$$

$$
\boxed{d\left(\frac{1}{S}\right) = \frac{1}{S}(\sigma^2 - \mu) \, dt - \frac{\sigma}{S} \, dB}
$$

Note the $\sigma^2$ term: the drift of $1/S$ has an additional $\sigma^2/S$ compared to the naive $-\mu/S$. This is the Ito correction for the reciprocal and is critical in computing quanto adjustments and exchange rate dynamics.

---

#### Problem 5d: Compute $d(e^{rt}S(t))$ and interpret

**Solution:** Using the Ito product rule with $f(t) = e^{rt}$ (deterministic) and $S(t)$ (stochastic):

$$
d(e^{rt}S) = e^{rt} \, dS + S \cdot r e^{rt} \, dt + d[e^{r\cdot}, S]_t
$$

Since $e^{rt}$ is deterministic (zero volatility), the cross-variation is zero:

$$
d(e^{rt}S) = e^{rt}(\mu S \, dt + \sigma S \, dB) + rSe^{rt} \, dt
$$

$$
= e^{rt}S[(\mu + r) \, dt + \sigma \, dB]
$$

**Interpretation:** $e^{rt}S(t)$ is the forward (future) value of the stock. Its growth rate is $\mu + r$.

> **Interview Tip:** Being fluent with these Ito's lemma calculations is essential. Practice until you can do them quickly and without error. Common mistakes: forgetting the $(dX)^2$ term, getting the sign wrong on $f''$, or forgetting that $(dB)^2 = dt$.

### Problem 6 (Bonus): Compute $\mathbb{E}[B(t)^2 B(s)]$ for $s < t$

**Solution:** Write $B(t) = B(s) + (B(t) - B(s))$ where $B(t) - B(s)$ is independent of $\mathcal{F}_s$.

$$
B(t)^2 = B(s)^2 + 2B(s)(B(t) - B(s)) + (B(t) - B(s))^2
$$

Multiply by $B(s)$:

$$
B(t)^2 B(s) = B(s)^3 + 2B(s)^2(B(t) - B(s)) + B(s)(B(t) - B(s))^2
$$

Taking expectations (using independence of $B(s)$ and $B(t) - B(s)$):

$$
\mathbb{E}[B(t)^2 B(s)] = \underbrace{\mathbb{E}[B(s)^3]}_{=0} + 2\underbrace{\mathbb{E}[B(s)^2]}_{=s} \cdot \underbrace{\mathbb{E}[B(t)-B(s)]}_{=0} + \underbrace{\mathbb{E}[B(s)]}_{=0} \cdot \underbrace{\mathbb{E}[(B(t)-B(s))^2]}_{=t-s}
$$

$$
\boxed{\mathbb{E}[B(t)^2 B(s)] = 0}
$$

This follows more quickly from the symmetry argument: since $B$ and $-B$ have the same distribution, $\mathbb{E}[B(t)^2 B(s)] = \mathbb{E}[B(t)^2 (-B(s))] = -\mathbb{E}[B(t)^2 B(s)]$, forcing the expectation to be zero.

### Problem 7 (Bonus): Compute $\mathbb{E}[e^{\int_0^T B(t) \, dt}]$

**Problem:** Find the expected value of $e^{\int_0^T B(t) \, dt}$.

**Solution:** Let $I = \int_0^T B(t) \, dt$. Since $B(t)$ is Gaussian and $I$ is a linear functional of the Gaussian process $B$, $I$ is also Gaussian. We need its mean and variance.

**Mean:**

$$
\mathbb{E}[I] = \int_0^T \mathbb{E}[B(t)] \, dt = 0
$$

**Variance:**

$$
\text{Var}(I) = \mathbb{E}[I^2] = \mathbb{E}\left[\int_0^T \int_0^T B(s) B(t) \, ds \, dt\right] = \int_0^T \int_0^T \min(s,t) \, ds \, dt
$$

$$
= 2\int_0^T \int_0^t s \, ds \, dt = 2 \int_0^T \frac{t^2}{2} \, dt = \int_0^T t^2 \, dt = \frac{T^3}{3}
$$

Since $I \sim \mathcal{N}(0, T^3/3)$ and $\mathbb{E}[e^{Z}] = e^{\text{Var}(Z)/2}$ for $Z \sim \mathcal{N}(0, \sigma^2)$:

$$
\boxed{\mathbb{E}\left[e^{\int_0^T B(t) \, dt}\right] = e^{T^3/6}}
$$

> **Interview Tip:** This problem tests whether you know that linear functionals of Gaussian processes are Gaussian. The key computation is the double integral $\int\int \min(s,t) \, ds \, dt = T^3/3$. This integral arises frequently in Asian option pricing (where the payoff depends on $\int_0^T S(t) \, dt$).
