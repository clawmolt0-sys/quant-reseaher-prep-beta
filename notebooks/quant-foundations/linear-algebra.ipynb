# Linear Algebra for Quantitative Finance

A comprehensive review of linear algebra concepts frequently tested in quant finance interviews. This notebook covers foundational theory, key results, and direct applications to portfolio optimization, factor models, and risk analysis.

---

## 1. Vectors and Vector Spaces

### 1.1 Vectors and Basic Definitions

A **vector** $\mathbf{v} \in \mathbb{R}^n$ is an ordered $n$-tuple of real numbers:

$$
\mathbf{v} = \begin{pmatrix} v_1 \\ v_2 \\ \vdots \\ v_n \end{pmatrix}
$$

A **vector space** $V$ over $\mathbb{R}$ is a set equipped with vector addition and scalar multiplication satisfying closure, associativity, commutativity, identity, and inverse axioms.

### 1.2 Span, Basis, and Dimension

- **Span**: The set of all linear combinations of a collection of vectors:

$$
\text{span}(\{\mathbf{v}_1, \ldots, \mathbf{v}_k\}) = \left\{ \sum_{i=1}^{k} \alpha_i \mathbf{v}_i \;\middle|\; \alpha_i \in \mathbb{R} \right\}
$$

- **Linear Independence**: Vectors $\{\mathbf{v}_1, \ldots, \mathbf{v}_k\}$ are linearly independent if

$$
\sum_{i=1}^k \alpha_i \mathbf{v}_i = \mathbf{0} \implies \alpha_1 = \alpha_2 = \cdots = \alpha_k = 0
$$

- **Basis**: A linearly independent set that spans $V$. Every basis of $V$ has the same number of elements.

- **Dimension**: $\dim(V)$ is the number of vectors in any basis of $V$.

### 1.3 Inner Products and Norms

The **standard inner product** (dot product) on $\mathbb{R}^n$:

$$
\langle \mathbf{u}, \mathbf{v} \rangle = \mathbf{u}^T \mathbf{v} = \sum_{i=1}^{n} u_i v_i
$$

**Norms** measure the "size" of a vector:

| Norm | Formula | Interpretation |
|------|---------|----------------|
| $L^2$ (Euclidean) | $\|\mathbf{x}\|_2 = \sqrt{\sum_{i} x_i^2}$ | Standard distance |
| $L^1$ (Manhattan) | $\|\mathbf{x}\|_1 = \sum_{i} |x_i|$ | Taxicab distance; promotes sparsity |
| $L^\infty$ (Max) | $\|\mathbf{x}\|_\infty = \max_i |x_i|$ | Largest absolute component |
| $L^p$ (General) | $\|\mathbf{x}\|_p = \left(\sum_{i} |x_i|^p\right)^{1/p}$ | Generalized norm |

### 1.4 Cauchy-Schwarz and Triangle Inequalities

**Cauchy-Schwarz inequality**:

$$
|\langle \mathbf{u}, \mathbf{v} \rangle| \leq \|\mathbf{u}\| \cdot \|\mathbf{v}\|
$$

Equality holds if and only if $\mathbf{u}$ and $\mathbf{v}$ are linearly dependent. This is the algebraic foundation of **correlation bounds**: $-1 \leq \rho_{XY} \leq 1$.

**Triangle inequality**:

$$
\|\mathbf{u} + \mathbf{v}\| \leq \|\mathbf{u}\| + \|\mathbf{v}\|
$$

### 1.5 Orthogonality and Projections

Two vectors are **orthogonal** if $\langle \mathbf{u}, \mathbf{v} \rangle = 0$. An **orthonormal** set satisfies $\langle \mathbf{e}_i, \mathbf{e}_j \rangle = \delta_{ij}$.

The **projection** of $\mathbf{u}$ onto $\mathbf{v}$:

$$
\text{proj}_{\mathbf{v}} \mathbf{u} = \frac{\langle \mathbf{u}, \mathbf{v} \rangle}{\langle \mathbf{v}, \mathbf{v} \rangle} \mathbf{v}
$$

The projection onto a subspace spanned by columns of $A$ (with $A$ having full column rank):

$$
P = A(A^T A)^{-1} A^T
$$

> **Interview Tip:** Projection matrices are idempotent ($P^2 = P$) and symmetric ($P^T = P$). The residual $\mathbf{u} - P\mathbf{u}$ is orthogonal to the column space of $A$. This is exactly the geometry behind OLS regression.

## 2. Matrices and Operations

### 2.1 Matrix Multiplication

For $A \in \mathbb{R}^{m \times n}$ and $B \in \mathbb{R}^{n \times p}$, the product $C = AB \in \mathbb{R}^{m \times p}$ has entries:

$$
C_{ij} = \sum_{k=1}^{n} A_{ik} B_{kj}
$$

Key properties:
- **Not commutative**: $AB \neq BA$ in general
- **Associative**: $(AB)C = A(BC)$
- **Distributive**: $A(B + C) = AB + AC$

### 2.2 Transpose

The transpose $A^T$ satisfies $(A^T)_{ij} = A_{ji}$. Key identities:

$$
(AB)^T = B^T A^T, \qquad (A^T)^T = A
$$

### 2.3 Trace

For a square matrix $A \in \mathbb{R}^{n \times n}$:

$$
\text{tr}(A) = \sum_{i=1}^{n} A_{ii}
$$

Essential trace identities:

| Identity | Condition |
|----------|-----------|
| $\text{tr}(A + B) = \text{tr}(A) + \text{tr}(B)$ | $A, B$ same size |
| $\text{tr}(cA) = c \cdot \text{tr}(A)$ | scalar $c$ |
| $\text{tr}(AB) = \text{tr}(BA)$ | $A \in \mathbb{R}^{m \times n}, B \in \mathbb{R}^{n \times m}$ |
| $\text{tr}(ABC) = \text{tr}(CAB) = \text{tr}(BCA)$ | cyclic permutation |
| $\text{tr}(A^T) = \text{tr}(A)$ | always |
| $\text{tr}(A) = \sum_{i} \lambda_i$ | eigenvalue sum |

The **trace trick** is widely used: $\mathbf{x}^T A \mathbf{x} = \text{tr}(\mathbf{x}^T A \mathbf{x}) = \text{tr}(A \mathbf{x} \mathbf{x}^T)$.

### 2.4 Special Matrices

| Matrix Type | Definition | Key Property |
|-------------|------------|--------------|
| **Symmetric** | $A = A^T$ | Real eigenvalues, orthogonal eigenvectors |
| **Orthogonal** | $Q^T Q = QQ^T = I$ | Preserves lengths and angles; $\det(Q) = \pm 1$ |
| **Diagonal** | $A_{ij} = 0$ for $i \neq j$ | Eigenvalues are the diagonal entries |
| **Idempotent** | $A^2 = A$ | Eigenvalues are 0 or 1; projection matrices |
| **Nilpotent** | $A^k = 0$ for some $k$ | All eigenvalues are 0 |
| **Positive Definite** | $\mathbf{x}^T A \mathbf{x} > 0$ for all $\mathbf{x} \neq \mathbf{0}$ | All eigenvalues $> 0$ |

### 2.5 Inverse

If $A$ is square and $\det(A) \neq 0$, then $A^{-1}$ exists with $AA^{-1} = A^{-1}A = I$.

Key identities:

$$
(AB)^{-1} = B^{-1}A^{-1}
$$

$$
(A^T)^{-1} = (A^{-1})^T
$$

$$
(cA)^{-1} = \frac{1}{c} A^{-1}
$$

**Sherman-Morrison formula** (rank-1 update of an inverse):

$$
(A + \mathbf{u}\mathbf{v}^T)^{-1} = A^{-1} - \frac{A^{-1}\mathbf{u}\mathbf{v}^T A^{-1}}{1 + \mathbf{v}^T A^{-1} \mathbf{u}}
$$

More generally, the **Woodbury identity**:

$$
(A + UCV)^{-1} = A^{-1} - A^{-1}U(C^{-1} + VA^{-1}U)^{-1}VA^{-1}
$$

> **Interview Tip:** The Sherman-Morrison and Woodbury formulas appear in quant interviews when discussing efficient updates to covariance matrix inverses (e.g., adding or removing an asset from a portfolio). Know the rank-1 case cold.

## 3. Systems of Linear Equations

### 3.1 The System $A\mathbf{x} = \mathbf{b}$

Given $A \in \mathbb{R}^{m \times n}$, $\mathbf{x} \in \mathbb{R}^n$, and $\mathbf{b} \in \mathbb{R}^m$:

- **Unique solution** exists iff $\text{rank}(A) = \text{rank}([A | \mathbf{b}]) = n$
- **Infinitely many solutions** if $\text{rank}(A) = \text{rank}([A | \mathbf{b}]) < n$
- **No solution** if $\text{rank}(A) < \text{rank}([A | \mathbf{b}])$

When $A$ is square and invertible: $\mathbf{x} = A^{-1}\mathbf{b}$.

### 3.2 Gaussian Elimination

The standard algorithm to solve $A\mathbf{x} = \mathbf{b}$:

1. Form the augmented matrix $[A | \mathbf{b}]$
2. Apply elementary row operations to reach **row echelon form** (REF)
3. Back-substitute to find $\mathbf{x}$

Elementary row operations:
- Swap two rows: $R_i \leftrightarrow R_j$
- Scale a row: $R_i \to cR_i$ (with $c \neq 0$)
- Add a multiple of one row to another: $R_i \to R_i + cR_j$

Computational complexity: $O(n^3)$ for an $n \times n$ system.

### 3.3 Rank and Nullity

- **Rank**: $\text{rank}(A)$ = number of linearly independent rows (or columns) = dimension of column space = dimension of row space.

- **Null space** (kernel): $\text{null}(A) = \{\mathbf{x} \in \mathbb{R}^n : A\mathbf{x} = \mathbf{0}\}$

- **Nullity**: $\text{nullity}(A) = \dim(\text{null}(A))$

### 3.4 The Rank-Nullity Theorem

For $A \in \mathbb{R}^{m \times n}$:

$$
\boxed{\text{rank}(A) + \text{nullity}(A) = n}
$$

This says: the dimension of the column space plus the dimension of the null space equals the number of columns.

### 3.5 The Four Fundamental Subspaces

For $A \in \mathbb{R}^{m \times n}$ with rank $r$:

| Subspace | Notation | Dimension | Lives in |
|----------|----------|-----------|----------|
| Column space | $\text{col}(A)$ | $r$ | $\mathbb{R}^m$ |
| Null space | $\text{null}(A)$ | $n - r$ | $\mathbb{R}^n$ |
| Row space | $\text{row}(A) = \text{col}(A^T)$ | $r$ | $\mathbb{R}^n$ |
| Left null space | $\text{null}(A^T)$ | $m - r$ | $\mathbb{R}^m$ |

Key orthogonality relations: $\text{col}(A) \perp \text{null}(A^T)$ and $\text{row}(A) \perp \text{null}(A)$.

> **Interview Tip:** Interviewers often ask: "If $A$ is $m \times n$ with $m < n$, can $A\mathbf{x} = \mathbf{0}$ have only the trivial solution?" Answer: No. By rank-nullity, $\text{nullity}(A) \geq n - m > 0$, so non-trivial solutions always exist.

## 4. Determinants

### 4.1 Definition and Computation

For a $2 \times 2$ matrix:

$$
\det\begin{pmatrix} a & b \\ c & d \end{pmatrix} = ad - bc
$$

For a $3 \times 3$ matrix, use **cofactor expansion** along row 1:

$$
\det(A) = a_{11}C_{11} + a_{12}C_{12} + a_{13}C_{13}
$$

where $C_{ij} = (-1)^{i+j} M_{ij}$ is the cofactor and $M_{ij}$ is the minor (determinant of the submatrix obtained by deleting row $i$ and column $j$).

### 4.2 Key Properties

| Property | Formula |
|----------|---------|
| Multiplicativity | $\det(AB) = \det(A)\det(B)$ |
| Transpose | $\det(A^T) = \det(A)$ |
| Inverse | $\det(A^{-1}) = \frac{1}{\det(A)}$ |
| Scalar multiple | $\det(cA) = c^n \det(A)$ for $A \in \mathbb{R}^{n \times n}$ |
| Row swap | Flips the sign of the determinant |
| Row scaling by $c$ | Multiplies the determinant by $c$ |
| Row addition | Does not change the determinant |
| Triangular matrix | $\det(A) = \prod_{i=1}^n a_{ii}$ |
| Singular matrix | $\det(A) = 0 \iff A$ is not invertible |

### 4.3 Geometric Interpretation

The absolute value $|\det(A)|$ gives the **volume scaling factor** of the linear transformation defined by $A$. If $\det(A) < 0$, the transformation reverses orientation.

### 4.4 Determinant and Eigenvalues

$$
\det(A) = \prod_{i=1}^{n} \lambda_i
$$

This connects invertibility to eigenvalues: $A$ is singular iff it has a zero eigenvalue.

> **Example:** Let $A = \begin{pmatrix} 3 & 1 \\ 1 & 3 \end{pmatrix}$. Find $\det(A)$.
>
> $\det(A) = 3 \cdot 3 - 1 \cdot 1 = 8$.
>
> The eigenvalues are $\lambda_1 = 4$, $\lambda_2 = 2$ (we can verify: $4 \times 2 = 8$ and $4 + 2 = 6 = \text{tr}(A)$).

## 5. Eigenvalues and Eigenvectors

### 5.1 Definition

A scalar $\lambda$ is an **eigenvalue** of $A \in \mathbb{R}^{n \times n}$ with corresponding **eigenvector** $\mathbf{v} \neq \mathbf{0}$ if:

$$
\boxed{A\mathbf{v} = \lambda \mathbf{v}}
$$

Rearranging: $(A - \lambda I)\mathbf{v} = \mathbf{0}$, so $\mathbf{v} \in \text{null}(A - \lambda I)$.

### 5.2 Characteristic Polynomial

Eigenvalues are the roots of the **characteristic polynomial**:

$$
p(\lambda) = \det(A - \lambda I) = 0
$$

For a $2 \times 2$ matrix $A = \begin{pmatrix} a & b \\ c & d \end{pmatrix}$:

$$
p(\lambda) = \lambda^2 - (a+d)\lambda + (ad - bc) = \lambda^2 - \text{tr}(A)\lambda + \det(A)
$$

### 5.3 Fundamental Properties

For $A \in \mathbb{R}^{n \times n}$ with eigenvalues $\lambda_1, \ldots, \lambda_n$ (counted with multiplicity):

$$
\text{tr}(A) = \sum_{i=1}^{n} \lambda_i, \qquad \det(A) = \prod_{i=1}^{n} \lambda_i
$$

Additional properties:

| If $\lambda$ is an eigenvalue of $A$... | Then... |
|----------------------------------------|---------|
| $A^k$ has eigenvalue | $\lambda^k$ |
| $A^{-1}$ has eigenvalue | $1/\lambda$ |
| $A + cI$ has eigenvalue | $\lambda + c$ |
| $cA$ has eigenvalue | $c\lambda$ |
| $A^T$ has eigenvalue | $\lambda$ (same eigenvalues) |

All of these share the same eigenvectors as $A$ (except $A^T$, which generally does not).

### 5.4 Diagonalization

A matrix $A$ is **diagonalizable** if there exists an invertible matrix $P$ and diagonal matrix $D$ such that:

$$
A = PDP^{-1}
$$

where the columns of $P$ are the eigenvectors of $A$ and the diagonal entries of $D$ are the corresponding eigenvalues.

**When is $A$ diagonalizable?**
- $A$ has $n$ linearly independent eigenvectors, OR equivalently,
- For each eigenvalue, the geometric multiplicity equals the algebraic multiplicity.

**Power of diagonalization**: $A^k = PD^kP^{-1}$, making computation of matrix powers trivial.

### 5.5 The Spectral Theorem

If $A$ is **real symmetric** ($A = A^T$), then:

1. All eigenvalues are **real**
2. Eigenvectors corresponding to distinct eigenvalues are **orthogonal**
3. $A$ is **orthogonally diagonalizable**: $A = Q\Lambda Q^T$ where $Q$ is orthogonal ($Q^T Q = I$)

The spectral decomposition can also be written as a sum of rank-1 matrices:

$$
A = \sum_{i=1}^{n} \lambda_i \mathbf{q}_i \mathbf{q}_i^T
$$

where $\mathbf{q}_i$ are the orthonormal eigenvectors.

> **Interview Tip:** The spectral theorem is the foundation for PCA, risk decomposition, and understanding covariance matrices. Covariance matrices are always real symmetric (and PSD), so the spectral theorem always applies to them.

### 5.6 Worked Example: Eigenvalues and Eigenvectors

> **Example:** Find the eigenvalues and eigenvectors of $A = \begin{pmatrix} 4 & 2 \\ 1 & 3 \end{pmatrix}$.
>
> **Step 1: Characteristic polynomial.**
>
> $$\det(A - \lambda I) = \det\begin{pmatrix} 4 - \lambda & 2 \\ 1 & 3 - \lambda \end{pmatrix} = (4 - \lambda)(3 - \lambda) - 2 = \lambda^2 - 7\lambda + 10 = 0$$
>
> **Step 2: Solve for eigenvalues.**
>
> $$(\lambda - 5)(\lambda - 2) = 0 \implies \lambda_1 = 5, \quad \lambda_2 = 2$$
>
> Quick check: $\text{tr}(A) = 7 = 5 + 2$, $\det(A) = 10 = 5 \times 2$.
>
> **Step 3: Find eigenvectors.**
>
> For $\lambda_1 = 5$: $(A - 5I)\mathbf{v} = \mathbf{0}$
>
> $$\begin{pmatrix} -1 & 2 \\ 1 & -2 \end{pmatrix} \mathbf{v} = \mathbf{0} \implies \mathbf{v}_1 = \begin{pmatrix} 2 \\ 1 \end{pmatrix}$$
>
> For $\lambda_2 = 2$: $(A - 2I)\mathbf{v} = \mathbf{0}$
>
> $$\begin{pmatrix} 2 & 2 \\ 1 & 1 \end{pmatrix} \mathbf{v} = \mathbf{0} \implies \mathbf{v}_2 = \begin{pmatrix} -1 \\ 1 \end{pmatrix}$$

## 6. Singular Value Decomposition (SVD)

### 6.1 The Decomposition

Every matrix $A \in \mathbb{R}^{m \times n}$ (not necessarily square) has a **singular value decomposition**:

$$
\boxed{A = U \Sigma V^T}
$$

where:
- $U \in \mathbb{R}^{m \times m}$ is orthogonal (columns are **left singular vectors**)
- $\Sigma \in \mathbb{R}^{m \times n}$ is diagonal with non-negative entries $\sigma_1 \geq \sigma_2 \geq \cdots \geq 0$ (the **singular values**)
- $V \in \mathbb{R}^{n \times n}$ is orthogonal (columns are **right singular vectors**)

### 6.2 Relationship to Eigendecomposition

The SVD is intimately connected to the eigendecompositions of $A^T A$ and $AA^T$:

- $A^T A = V \Sigma^T \Sigma V^T$ (eigendecomposition of $A^T A$)
- $AA^T = U \Sigma \Sigma^T U^T$ (eigendecomposition of $AA^T$)

The singular values of $A$ are the **square roots of the eigenvalues** of $A^T A$ (or $AA^T$):

$$
\sigma_i = \sqrt{\lambda_i(A^T A)}
$$

### 6.3 Properties of Singular Values

- $\text{rank}(A) =$ number of non-zero singular values
- $\|A\|_F = \sqrt{\sum_{i} \sigma_i^2}$ (Frobenius norm)
- $\|A\|_2 = \sigma_1$ (spectral norm / operator norm)
- $\text{condition number:} \; \kappa(A) = \sigma_1 / \sigma_r$ (where $r = \text{rank}(A)$)

### 6.4 Low-Rank Approximation (Eckart-Young Theorem)

The best rank-$k$ approximation to $A$ (in both Frobenius and spectral norms) is:

$$
A_k = \sum_{i=1}^{k} \sigma_i \mathbf{u}_i \mathbf{v}_i^T
$$

The approximation error:

$$
\|A - A_k\|_F = \sqrt{\sum_{i=k+1}^{r} \sigma_i^2}
$$

### 6.5 The Pseudoinverse

The **Moore-Penrose pseudoinverse** is:

$$
A^+ = V \Sigma^+ U^T
$$

where $\Sigma^+$ is formed by taking the reciprocal of each non-zero entry of $\Sigma$ (and transposing).

For the overdetermined system $A\mathbf{x} = \mathbf{b}$ (no exact solution), the **least squares solution** is:

$$
\mathbf{x}^* = A^+ \mathbf{b}
$$

> **Interview Tip:** SVD is one of the most powerful tools in applied mathematics. In finance, it is used for dimensionality reduction of return matrices, computing pseudoinverses for underdetermined regression, and detecting near-singular covariance matrices. If an interviewer asks about numerical stability, SVD is almost always the right answer.

## 7. Positive Definite Matrices

### 7.1 Definitions

A symmetric matrix $A \in \mathbb{R}^{n \times n}$ is:

- **Positive definite (PD)** if $\mathbf{x}^T A \mathbf{x} > 0$ for all $\mathbf{x} \neq \mathbf{0}$
- **Positive semi-definite (PSD)** if $\mathbf{x}^T A \mathbf{x} \geq 0$ for all $\mathbf{x}$
- **Negative definite** if $\mathbf{x}^T A \mathbf{x} < 0$ for all $\mathbf{x} \neq \mathbf{0}$
- **Indefinite** if $\mathbf{x}^T A \mathbf{x}$ can be positive or negative

### 7.2 Equivalent Conditions for Positive Definiteness

For a symmetric matrix $A$, the following are all equivalent:

1. $\mathbf{x}^T A \mathbf{x} > 0$ for all $\mathbf{x} \neq \mathbf{0}$
2. All eigenvalues of $A$ are strictly positive: $\lambda_i > 0$ for all $i$
3. All leading principal minors (determinants of top-left $k \times k$ submatrices) are positive
4. $A$ has a **Cholesky decomposition**: $A = LL^T$ where $L$ is lower triangular with positive diagonal
5. There exists an invertible matrix $B$ such that $A = B^T B$
6. All pivots in Gaussian elimination (without row swaps) are positive

### 7.3 Cholesky Decomposition

If $A$ is symmetric and positive definite, it can be uniquely factored as:

$$
A = LL^T
$$

where $L$ is lower triangular with positive diagonal entries. This is the "square root" of a positive definite matrix.

**Why it matters in finance**: Cholesky decomposition is the standard method for generating **correlated random variables** in Monte Carlo simulation. Given a correlation matrix $\Sigma$:

1. Compute $L$ where $\Sigma = LL^T$
2. Generate independent standard normals $\mathbf{z} \sim \mathcal{N}(\mathbf{0}, I)$
3. Set $\mathbf{x} = L\mathbf{z}$, then $\mathbf{x} \sim \mathcal{N}(\mathbf{0}, \Sigma)$

### 7.4 Covariance Matrices Are PSD

The covariance matrix of any random vector is always positive semi-definite. If $\mathbf{X}$ is a random vector:

$$
\Sigma = \text{Cov}(\mathbf{X}) = E[(\mathbf{X} - \boldsymbol{\mu})(\mathbf{X} - \boldsymbol{\mu})^T]
$$

Proof of PSD: For any $\mathbf{a}$,

$$
\mathbf{a}^T \Sigma \mathbf{a} = \text{Var}(\mathbf{a}^T \mathbf{X}) \geq 0
$$

The covariance matrix is strictly PD iff no non-trivial linear combination of the components has zero variance (i.e., no component is a perfect linear function of the others).

> **Interview Tip:** A very common interview question: "How do you know a given matrix is a valid covariance matrix?" Answer: It must be symmetric and positive semi-definite. If a sample covariance matrix loses PD due to estimation error, you can fix it by setting negative eigenvalues to a small positive number (spectral clipping) or using shrinkage estimators.

## 8. Principal Component Analysis (PCA)

### 8.1 Setup

Given a centered data matrix $X \in \mathbb{R}^{n \times p}$ (each row is an observation, each column a variable), the **sample covariance matrix** is:

$$
\Sigma = \frac{1}{n-1} X^T X
$$

(Using $n-1$ for the unbiased estimator; some formulations use $\frac{1}{n}$.)

### 8.2 PCA via Eigendecomposition

Compute the eigendecomposition of the covariance matrix:

$$
\Sigma = Q \Lambda Q^T
$$

where $\Lambda = \text{diag}(\lambda_1, \ldots, \lambda_p)$ with $\lambda_1 \geq \lambda_2 \geq \cdots \geq \lambda_p \geq 0$.

- **Principal components**: The columns $\mathbf{q}_1, \ldots, \mathbf{q}_p$ of $Q$ (eigenvectors)
- **$k$-th principal component scores**: $\mathbf{z}_k = X \mathbf{q}_k \in \mathbb{R}^n$
- **Variance explained** by the $k$-th component: $\lambda_k$

### 8.3 Variance Explained

The **fraction of total variance** explained by the first $k$ components:

$$
\text{FVE}(k) = \frac{\sum_{i=1}^{k} \lambda_i}{\sum_{i=1}^{p} \lambda_i} = \frac{\sum_{i=1}^{k} \lambda_i}{\text{tr}(\Sigma)}
$$

A common heuristic: choose $k$ such that $\text{FVE}(k) \geq 0.90$ or $0.95$.

### 8.4 PCA via SVD

Equivalently, compute the SVD of the centered data matrix:

$$
X = U \Sigma_X V^T
$$

Then the columns of $V$ are the principal component directions, and the principal component scores are the columns of $U\Sigma_X$. This is often more numerically stable than forming $X^TX$ explicitly.

### 8.5 Dimensionality Reduction

Project data onto the first $k$ principal components:

$$
X_{\text{reduced}} = X Q_k
$$

where $Q_k \in \mathbb{R}^{p \times k}$ contains the first $k$ eigenvectors. This gives the best $k$-dimensional representation in the least-squares sense.

### 8.6 Applications in Finance

**Yield curve modeling:** PCA on historical yield changes typically produces:
- **PC1** ($\sim 85\%$ of variance): Level (parallel shift)
- **PC2** ($\sim 10\%$ of variance): Slope (steepening/flattening)
- **PC3** ($\sim 3\%$ of variance): Curvature (butterfly)

**Equity returns:**
- The first principal component often corresponds to the **market factor**
- Subsequent PCs may capture sector, size, or value/growth effects
- PCA provides a data-driven approach to factor modeling: $\mathbf{r}_t \approx B \mathbf{f}_t$

**Risk management:**
- Reduce dimensionality of large covariance matrices
- Identify dominant risk factors
- Regularize covariance estimation

> **Interview Tip:** PCA on yield curve data is an extremely common interview topic. Be prepared to explain what the first three PCs represent and why the first PC captures the vast majority of variance. The key insight: interest rates across maturities are highly correlated, so most variation is captured by a few factors.

## 9. Matrix Calculus

### 9.1 Gradient with Respect to a Vector

The gradient of a scalar function $f(\mathbf{x})$ with respect to $\mathbf{x} \in \mathbb{R}^n$ is:

$$
\nabla_{\mathbf{x}} f = \begin{pmatrix} \frac{\partial f}{\partial x_1} \\ \vdots \\ \frac{\partial f}{\partial x_n} \end{pmatrix}
$$

### 9.2 Key Results

| Function $f(\mathbf{x})$ | Gradient $\nabla_{\mathbf{x}} f$ |
|--------------------------|----------------------------------|
| $\mathbf{a}^T \mathbf{x}$ | $\mathbf{a}$ |
| $\mathbf{x}^T \mathbf{x}$ | $2\mathbf{x}$ |
| $\mathbf{x}^T A \mathbf{x}$ | $(A + A^T)\mathbf{x}$ |
| $\mathbf{x}^T A \mathbf{x}$ (symmetric $A$) | $2A\mathbf{x}$ |
| $\|A\mathbf{x} - \mathbf{b}\|_2^2$ | $2A^T(A\mathbf{x} - \mathbf{b})$ |

### 9.3 Derivation of the Quadratic Form Gradient

For $f(\mathbf{x}) = \mathbf{x}^T A \mathbf{x} = \sum_{i,j} A_{ij} x_i x_j$:

$$
\frac{\partial f}{\partial x_k} = \sum_j A_{kj} x_j + \sum_i A_{ik} x_i = (A\mathbf{x})_k + (A^T\mathbf{x})_k
$$

Therefore $\nabla_{\mathbf{x}} (\mathbf{x}^T A \mathbf{x}) = (A + A^T)\mathbf{x}$.

When $A$ is symmetric ($A = A^T$), this simplifies to $2A\mathbf{x}$.

### 9.4 The Hessian

The Hessian of $f(\mathbf{x})$ is the matrix of second partial derivatives:

$$
H_f = \nabla^2 f = \begin{pmatrix} \frac{\partial^2 f}{\partial x_1^2} & \cdots & \frac{\partial^2 f}{\partial x_1 \partial x_n} \\ \vdots & \ddots & \vdots \\ \frac{\partial^2 f}{\partial x_n \partial x_1} & \cdots & \frac{\partial^2 f}{\partial x_n^2} \end{pmatrix}
$$

For the quadratic form $f(\mathbf{x}) = \mathbf{x}^T A \mathbf{x}$ with symmetric $A$:

$$
\nabla^2 f = 2A
$$

The function is convex iff the Hessian is PSD, i.e., iff $A$ is PSD.

### 9.5 Application: Portfolio Optimization

**Minimum-variance portfolio** (no short-selling constraints):

$$
\min_{\mathbf{w}} \; \mathbf{w}^T \Sigma \mathbf{w} \quad \text{subject to} \quad \mathbf{1}^T \mathbf{w} = 1
$$

Using the Lagrangian:

$$
\mathcal{L}(\mathbf{w}, \lambda) = \mathbf{w}^T \Sigma \mathbf{w} - \lambda(\mathbf{1}^T \mathbf{w} - 1)
$$

Setting $\nabla_{\mathbf{w}} \mathcal{L} = 2\Sigma \mathbf{w} - \lambda \mathbf{1} = \mathbf{0}$:

$$
\mathbf{w} = \frac{\lambda}{2} \Sigma^{-1} \mathbf{1}
$$

Applying the constraint $\mathbf{1}^T \mathbf{w} = 1$:

$$
\boxed{\mathbf{w}^* = \frac{\Sigma^{-1} \mathbf{1}}{\mathbf{1}^T \Sigma^{-1} \mathbf{1}}}
$$

> **Interview Tip:** Be prepared to derive the minimum-variance portfolio from scratch using matrix calculus and Lagrange multipliers. This is one of the most common quant interview derivations. Also know the mean-variance optimal portfolio when a target return constraint is added (the two-fund theorem).

## 10. Applications in Quantitative Finance

### 10.1 Mean-Variance Portfolio Optimization

Given $n$ assets with expected return vector $\boldsymbol{\mu} \in \mathbb{R}^n$ and covariance matrix $\Sigma \in \mathbb{R}^{n \times n}$.

**Maximum Sharpe ratio portfolio** (tangency portfolio):

$$
\mathbf{w}^* = \frac{\Sigma^{-1}(\boldsymbol{\mu} - r_f \mathbf{1})}{\mathbf{1}^T \Sigma^{-1}(\boldsymbol{\mu} - r_f \mathbf{1})}
$$

**Mean-variance problem** with target return $\mu_p$:

$$
\min_{\mathbf{w}} \; \mathbf{w}^T \Sigma \mathbf{w} \quad \text{s.t.} \quad \boldsymbol{\mu}^T \mathbf{w} = \mu_p, \; \mathbf{1}^T \mathbf{w} = 1
$$

The solution lies on the **efficient frontier**, a parabola in $(\sigma_p, \mu_p)$ space.

### 10.2 Factor Models

A **linear factor model** for asset returns:

$$
\mathbf{r} = \boldsymbol{\alpha} + B \mathbf{f} + \boldsymbol{\epsilon}
$$

where:
- $\mathbf{r} \in \mathbb{R}^n$: asset returns
- $\mathbf{f} \in \mathbb{R}^k$: factor returns ($k \ll n$)
- $B \in \mathbb{R}^{n \times k}$: factor loading matrix (betas)
- $\boldsymbol{\epsilon} \in \mathbb{R}^n$: idiosyncratic noise, $\text{Cov}(\boldsymbol{\epsilon}) = D$ (diagonal)

The covariance structure becomes:

$$
\Sigma = B \Sigma_f B^T + D
$$

This reduces the number of parameters from $O(n^2)$ to $O(nk)$, which is critical for large portfolios.

### 10.3 Ordinary Least Squares (OLS) Regression

The linear regression problem $\mathbf{y} = X\boldsymbol{\beta} + \boldsymbol{\epsilon}$ has the OLS solution:

$$
\boxed{\hat{\boldsymbol{\beta}} = (X^T X)^{-1} X^T \mathbf{y}}
$$

**Derivation**: Minimize $\|\mathbf{y} - X\boldsymbol{\beta}\|^2$:

$$
\nabla_{\boldsymbol{\beta}} (\mathbf{y} - X\boldsymbol{\beta})^T(\mathbf{y} - X\boldsymbol{\beta}) = -2X^T(\mathbf{y} - X\boldsymbol{\beta}) = \mathbf{0}
$$

$$
\implies X^T X \boldsymbol{\beta} = X^T \mathbf{y} \implies \hat{\boldsymbol{\beta}} = (X^T X)^{-1} X^T \mathbf{y}
$$

Properties:
- Projection matrix (hat matrix): $H = X(X^TX)^{-1}X^T$
- Fitted values: $\hat{\mathbf{y}} = H\mathbf{y}$
- Residuals: $\hat{\boldsymbol{\epsilon}} = (I - H)\mathbf{y}$
- $\text{Cov}(\hat{\boldsymbol{\beta}}) = \sigma^2 (X^TX)^{-1}$

### 10.4 Risk Decomposition via Eigenvalues

The portfolio variance is:

$$
\sigma_p^2 = \mathbf{w}^T \Sigma \mathbf{w} = \mathbf{w}^T \left(\sum_{i=1}^{n} \lambda_i \mathbf{q}_i \mathbf{q}_i^T\right) \mathbf{w} = \sum_{i=1}^{n} \lambda_i (\mathbf{q}_i^T \mathbf{w})^2
$$

This shows that portfolio risk decomposes into contributions from each **principal component**, weighted by the eigenvalue (variance along that component) and the portfolio's exposure $(\mathbf{q}_i^T \mathbf{w})^2$ to that component.

### 10.5 Correlation Matrices: Properties and Challenges

A valid **correlation matrix** must be:
- Symmetric
- All diagonal entries equal to 1
- Positive semi-definite
- All off-diagonal entries in $[-1, 1]$

**Common challenge:** Not all matrices with these apparent properties are valid correlation matrices. For example, consider three assets with pairwise correlations $\rho_{12} = 0.9$, $\rho_{13} = 0.9$, $\rho_{23} = -0.9$. The resulting matrix:

$$
C = \begin{pmatrix} 1 & 0.9 & 0.9 \\ 0.9 & 1 & -0.9 \\ 0.9 & -0.9 & 1 \end{pmatrix}
$$

has $\det(C) = 1 + 2(0.9)(0.9)(-0.9) - 0.81 - 0.81 - 0.81 = -2.888 < 0$, so it has a negative eigenvalue and is **not** a valid correlation matrix.

> **Interview Tip:** A common question is: "Can three assets have pairwise correlations of 0.9, 0.9, and -0.9?" The answer is no, because the resulting correlation matrix is not PSD. Always check via the determinant or eigenvalues.

## 11. Classic Interview Problems

### Problem 1: Powers of a Matrix via Diagonalization

> **Problem:** Let $A = \begin{pmatrix} 2 & 1 \\ 0 & 3 \end{pmatrix}$. Compute $A^{100}$.

**Solution:**

**Step 1: Find eigenvalues.** Since $A$ is upper triangular, the eigenvalues are the diagonal entries: $\lambda_1 = 2$, $\lambda_2 = 3$.

**Step 2: Find eigenvectors.**

For $\lambda_1 = 2$: $(A - 2I)\mathbf{v} = \begin{pmatrix} 0 & 1 \\ 0 & 1 \end{pmatrix}\mathbf{v} = \mathbf{0} \implies \mathbf{v}_1 = \begin{pmatrix} 1 \\ 0 \end{pmatrix}$

For $\lambda_2 = 3$: $(A - 3I)\mathbf{v} = \begin{pmatrix} -1 & 1 \\ 0 & 0 \end{pmatrix}\mathbf{v} = \mathbf{0} \implies \mathbf{v}_2 = \begin{pmatrix} 1 \\ 1 \end{pmatrix}$

**Step 3: Diagonalize.** $A = PDP^{-1}$ where

$$
P = \begin{pmatrix} 1 & 1 \\ 0 & 1 \end{pmatrix}, \quad D = \begin{pmatrix} 2 & 0 \\ 0 & 3 \end{pmatrix}, \quad P^{-1} = \begin{pmatrix} 1 & -1 \\ 0 & 1 \end{pmatrix}
$$

**Step 4: Compute.** $A^{100} = PD^{100}P^{-1}$:

$$
A^{100} = \begin{pmatrix} 1 & 1 \\ 0 & 1 \end{pmatrix} \begin{pmatrix} 2^{100} & 0 \\ 0 & 3^{100} \end{pmatrix} \begin{pmatrix} 1 & -1 \\ 0 & 1 \end{pmatrix} = \begin{pmatrix} 2^{100} & 3^{100} - 2^{100} \\ 0 & 3^{100} \end{pmatrix}
$$

### Problem 2: Positive Definiteness of a Correlation Matrix

> **Problem:** Determine the range of $\rho$ for which the following is a valid correlation matrix:
>
> $$C = \begin{pmatrix} 1 & 0.5 & \rho \\ 0.5 & 1 & 0.5 \\ \rho & 0.5 & 1 \end{pmatrix}$$

**Solution:**

We need $C$ to be PSD. Check via the determinant (all leading principal minors must be non-negative).

- $1 \times 1$ minor: $1 > 0$ \checkmark
- $2 \times 2$ minor: $1 - 0.25 = 0.75 > 0$ \checkmark
- $3 \times 3$ determinant:

$$
\det(C) = 1(1 - 0.25) - 0.5(0.5 - 0.5\rho) + \rho(0.25 - \rho)
$$

$$
= 0.75 - 0.25 + 0.25\rho + 0.25\rho - \rho^2
$$

$$
= -\rho^2 + 0.5\rho + 0.5
$$

Setting $\det(C) \geq 0$:

$$
\rho^2 - 0.5\rho - 0.5 \leq 0
$$

$$
\rho = \frac{0.5 \pm \sqrt{0.25 + 2}}{2} = \frac{0.5 \pm 1.5}{2}
$$

$$
\rho \in [-0.5, \; 1]
$$

Combined with the requirement $\rho \in [-1, 1]$ for a correlation, the answer is $\rho \in [-0.5, \; 1]$.

> **Interview Tip:** This type of problem tests whether you understand that not all "reasonable-looking" correlations can coexist. The constraint is positive semi-definiteness of the correlation matrix.

### Problem 3: Deriving OLS via Projection

> **Problem:** Show that the OLS solution $\hat{\boldsymbol{\beta}} = (X^TX)^{-1}X^T\mathbf{y}$ is the projection of $\mathbf{y}$ onto the column space of $X$.

**Solution:**

We want to find $\hat{\mathbf{y}} = X\hat{\boldsymbol{\beta}}$ that is the closest point in $\text{col}(X)$ to $\mathbf{y}$.

**Step 1:** The residual must be orthogonal to the column space of $X$:

$$
X^T(\mathbf{y} - X\hat{\boldsymbol{\beta}}) = \mathbf{0}
$$

**Step 2:** Expand and solve the **normal equations**:

$$
X^T \mathbf{y} = X^T X \hat{\boldsymbol{\beta}}
$$

$$
\hat{\boldsymbol{\beta}} = (X^TX)^{-1}X^T\mathbf{y}
$$

**Step 3:** The projection of $\mathbf{y}$ onto $\text{col}(X)$ is:

$$
\hat{\mathbf{y}} = X\hat{\boldsymbol{\beta}} = X(X^TX)^{-1}X^T \mathbf{y} = P\mathbf{y}
$$

where $P = X(X^TX)^{-1}X^T$ is the projection matrix.

**Verify it is a projection:** $P^2 = X(X^TX)^{-1}X^TX(X^TX)^{-1}X^T = X(X^TX)^{-1}X^T = P$ \checkmark

**Verify symmetry:** $P^T = (X(X^TX)^{-1}X^T)^T = X((X^TX)^{-1})^TX^T = X(X^TX)^{-1}X^T = P$ \checkmark

The geometric interpretation: OLS finds the point in the column space of $X$ that is closest to $\mathbf{y}$ by dropping a perpendicular from $\mathbf{y}$ to $\text{col}(X)$.

### Problem 4: Eigenvalue Perturbation and Risk

> **Problem:** A portfolio of 3 assets has covariance matrix
>
> $$\Sigma = \begin{pmatrix} 4 & 2 & 0 \\ 2 & 5 & 1 \\ 0 & 1 & 2 \end{pmatrix}$$
>
> For the equally-weighted portfolio $\mathbf{w} = (1/3, 1/3, 1/3)^T$, compute the portfolio variance and decompose it by principal components.

**Solution:**

**Step 1: Portfolio variance.**

$$
\sigma_p^2 = \mathbf{w}^T \Sigma \mathbf{w} = \frac{1}{9} \mathbf{1}^T \Sigma \mathbf{1}
$$

$$
\Sigma \mathbf{1} = \begin{pmatrix} 6 \\ 8 \\ 3 \end{pmatrix}, \quad \mathbf{1}^T \Sigma \mathbf{1} = 6 + 8 + 3 = 17
$$

$$
\sigma_p^2 = \frac{17}{9} \approx 1.889
$$

**Step 2: Eigendecomposition.** The eigenvalues of $\Sigma$ are approximately $\lambda_1 \approx 6.29$, $\lambda_2 \approx 3.37$, $\lambda_3 \approx 1.34$ (these sum to $\text{tr}(\Sigma) = 11$).

**Step 3: Risk decomposition.** The portfolio variance decomposes as:

$$
\sigma_p^2 = \sum_{i=1}^3 \lambda_i (\mathbf{q}_i^T \mathbf{w})^2
$$

The first PC (associated with $\lambda_1 \approx 6.29$) captures the largest source of risk. This tells us which principal component contributes most to the portfolio's total variance, enabling targeted risk management.

> **Interview Tip:** In practice, if one eigenvalue dominates, the portfolio risk is essentially one-dimensional. This is why PCA-based risk models can dramatically simplify risk analysis for large portfolios.
